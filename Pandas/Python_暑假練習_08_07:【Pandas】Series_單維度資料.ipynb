```python
#===========================================
# 1. Series 與 「自訂索引 (Index)」
#===========================================
import pandas as pd
print("--------(Pandas: Series 自訂索引)--------")

# 純 Python 陣列中，資料只能用 0, 1, 2...來抓取
# 但在 Series 中，可以為每筆資料貼上「專屬索引標籤 (Index)」

# 假設量測了四顆電阻的實際阻值 (單位: Ohm)
resistances= pd.Series(
[99.8, 100.2, 99.5, 100.5],index= ["R1", "R2", "R3", "R4"]
)

print(f"【電阻量測數據】\n{resistances}\n")

# 透過自訂索引抓取資料，程式碼會變得非常直覺:
print(f"抓取單顆電阻\n{resistances['R2']}\n")
print(f"抓取多顆電阻\n{resistances[['R1', 'R4']]}\n")


#§補充資料:
# 在 Pandas 中，想一次抓取多個資料時，必須傳入一個「列表 (List)」
# 外層的中括號 []： 代表「要開始抓資料了」
# 內層的中括號 []： 代表「一個包含多個標籤的列表」


#======================================
# 2. 觀察 Series 的資料屬性
#======================================
print("\n--------(觀察 Series 的資料屬性)--------")

print(f"資料型態 (.dtype): {resistances.dtype}")
print(f"純數值陣列 (.values): {resistances.values}")
print(f"所有索引標籤 (.index): {resistances.index}")


#============================
# 向量化運算 (廣播機制)
#============================
print("\n--------(Series 的數學運算)--------")

# 假設這四顆電阻都串聯，且流過 0.05A (50mA) 的電流
# 只要一行程式碼就能算出所有電阻兩端的跨壓 (V= I*R)
# 不用寫 for 迴圈，Pandas 會自動把 0.05 乘到每一筆資料上
resistances= pd.Series(
[99.8, 100.2, 99.5, 100.5],index= ["R1", "R2", "R3", "R4"]
)

current= 0.05
voltages= resistances * current

print(f"【各電阻的跨壓 (V)】\n{voltages}\n")


#===========================================================
# Pandas 的強大功能: 自動索引對齊 (Index Alignment)
#===========================================================
print("--------(Series 自動索引對齊)--------")

# 這是 Series 最強大的功能
# 假設用另一台儀器量測了各電阻的電流，但「順序是打亂的」


# 範例: 這裡故意遺漏 R3，並且多了一個 R5
measured_currents= pd.Series(
[0.05, 0.04, 0.06, 0.05],index= ["R2", "R1", "R5", "R4"]
)

resistances= pd.Series(
[99.8, 100.2, 99.5, 100.5],index= ["R1", "R2", "R3", "R4"]
)

# 直接拿兩個 Series 相乘計算功率 (P= I**2 * R)
# Pandas 會「自動比對索引標籤」，只計算索引標籤對得上的資料，不會因為順序錯亂而算錯
power= measured_currents**2 * resistances
print(f"【各電阻的消耗功率 (W)】\n{power}\n")


#§觀察輸出結果:
# 發現 R3 與 R5 計算出來的數值變成了 NaN (Not a Number，缺失值)
# 因為它們沒有成對的資料可以做計算，所以 Pandas 自動標記為缺失
```
