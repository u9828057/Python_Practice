```python
# def: 函式
print("--------(def: 函式的定義與呼叫)--------")

#=================================
#    1. 基本語法:
#
#  def 函式名稱(參數名稱):
#    函式內部的程式碼
#=================================

# 定義函式 (Define Function): 就像在打造一個專屬工具箱，定義時不會執行裡面的程式碼
def greet_user(name):
  print(f"歡迎光臨，{name} 歡迎來到 Python 的世界!")

# 呼叫函式 (Call Function): 呼叫它的名字並傳入資料，工具箱才會真的啟動
greet_user("工程師")
greet_user("Python Master")


#======================================
#  2. print() vs return 的差異
#======================================
print("\n--------(print vs. return)--------")

# 只用 print (僅在螢幕上顯示，但電腦沒有紀錄資料、結果)
def add_with_print(a,b):
  print(a+b)

# 使用 return (把計算後的資料「交還」給程式，讓外面的變數可以抓取繼續使用)
def add_with_return(a,b):
  total=a+b
  return total

# 兩者的差異測試
result_print=add_with_print(3,5)      #螢幕會印出 8，但 result_print 收到的數值是None
result_return=add_with_return(3,5)     #螢幕沒有畫面，但數值 8 已經確實存入 result_return

print(f"用 print 儲存的變數內容: {result_print}")         #輸出: None(空值)
print(f"用 return 儲存的變數內容: {result_return}")        #輸出: 8(拿到return儲存後的資料)
print(f"將 return 的結果拿來做二次運算: {result_return*10}")   #輸出: 80


#=======================================================
# 3. 參數的進階用法: 預設參數(Default Arguments)
#=======================================================
print("\n--------(3. 預設參數的用法)--------")

# 範例: 計算電功率，若無特別填寫電阻值，預設電阻值 R=10 Ohm
def calculate_power(voltage, resistance=10):
  power=(voltage*voltage) / resistance
  return power

# 呼叫時不填入第二個參數，自動採用預設值 R=10
p1=calculate_power(20)
print(f"使用預設電阻的參數: {p1}W")

# 呼叫時填入第二個參數，複寫預設值，採用 R=5
p2=calculate_power(20,5)
print(f"使用自訂電阻值的功率: {p2}W")
# 💡：有預設值的參數，在定義時「必須」放在沒有預設值的參數後面


# 【return 的替換魔法】
# return 的核心作用：把算出來的「結果數值」，端回到「當時呼叫這個函式的位置」，
# 並且直接「替換」掉原本的那串呼叫指令！
#
# Example: p1=calculate_power(20)
# 以這行代碼為例，電腦內部的執行動作如下：
# 步驟1: 電腦看到 calculate_power(20)，帶著 20 進入函式 power=(20*20)/10 計算出結果 40.0
# 步驟2: 遇到 return，電腦把 calculate_power(20) 這整串指令直接「一秒替換」成數字 40.0
# 步驟3: 因此這行代碼在電腦眼裡瞬間變成了：p1 = 40.0，最後成功把 40.0 存進變數 p1 裡！


#========================================
# 4. 回傳多個資料: 與 Tuple 結合
#========================================
print("\n--------(4. 回傳多個資料與 Tuple 拆解)--------")

# 當需要一次處理並回傳多個數據 (如:波形量測同時找出最大值與最小值)
signals=[12,60,15,70,90,5]
def analyze_signals(data_list):
  max_value=max(data_list)
  min_value=min(data_list)
  length=len(data_list)

# 用逗號分隔多個回傳值，Python 會自動將它們打包成一個 Tuple(元組) 回傳
  return max_value, min_value, length

# 運用 Tuple技巧: 多變數拆解(Unpacking)
highest, lowest, count=analyze_signals(signals)

print(f"訊號總數: {count}筆")
print(f"最高電壓: {highest}V")
print(f"最低電壓: {lowest}V")


#========================================================
# 5. 變數的生存範圍 (Scope:全域變數 vs 區域變數)
#========================================================
print("\n--------(5. 變數的生存範圍: 全域變數 vs 區域變數)--------")

global_var="我是全域變數 (大家都能用)"           #全域變數放在函式外面
def test_scope():
  local_var="我是區域變數 (只有函式內部看得見)"     #區域變數放在函式裡面
  print(f"在函式內部讀取: {global_var}")
  print(f"在函式內部讀取: {local_var}")

test_scope()                #呼叫函式 test_scope()，函式內部的程式碼跑了一次，所以出現兩個print的內容
print(f"在函式外部讀取: {global_var}")
#print(f"在函式外部讀取: {local_var}")  #輸出錯誤: 在外面找不到 "local_var" 的變數名稱"


#=====================================================================
# 5-1. 進階補充：如何在函式內部「修改」全域變數？(global 關鍵字)
#=====================================================================
print("\n--------(5-1. 進階補充：global)--------")

player_score=0   #建立一個全域變數

def add_score():
  global player_score   #在變數前加上 global ，宣告「我要修改的是外面的變數: player_score」
  player_score=100     #如果沒寫上面那行 global player_score，直接寫這行，會被當成創立新的區域變數
  print(f"在函式內部修改了分數，目前分數為: {player_score}")

print(f"呼叫函式之前，原本的全域分數為: {player_score}")  #函式跑過一次前的全域變數: player_score=0
add_score()                         #呼叫函式，跑過一次函式內部的程式碼
print(f"呼叫函式之後，全域分數為: {player_score}")     #函式跑過一次後的全域變數: player_score=100


#=============================================================
# 6. 關鍵字參數 (Keyword Arguments): 不需記得填寫順序
#=============================================================
print("\n--------(6. 關鍵字參數)--------")

def calculate_power(voltage,resistance=10):
  power=(voltage*voltage) / resistance
  return power

# 傳統位置參數: 必須按照順序填寫(先 voltage， 再 resistance)
print(f"位置參數結果: {calculate_power(20,5)}W")

# 關鍵字參數: 直接填入參數名稱與數值，順序顛倒也正確
print(f"關鍵字參數結果: {calculate_power(resistance=5, voltage=20)}W")


#====================================================
# 7. 不定數量參數 (*args): 接收任意數量的輸入
#====================================================
print("\n--------(7. 不定數量參數 *args)--------")

# 參數前面加 * ，Python 會把收到的所有數值自動打包成一個 Tuple(元組)
def calculate_average_voltage(*voltage):
  #【防呆機制】如果收到的 voltage 是空的 (沒有傳入任何參數)
#  if not voltage:
#    return 0.0, 0         #直接安全的預設值 (平均為0.0V，筆數為0)
  total=sum(voltage)
  count=len(voltage)
  return total/count, count     #回傳 total/count與count 兩個數值

avg1,c1= calculate_average_voltage(12,15,18)    #用 avg1與c1 兩個變數接住函式回傳的兩個數值
avg2,c2= calculate_average_voltage(5,10,15,20,25,30)

print(f"{c1} 筆電壓的平均值: {avg1:.1f}V")
print(f"{c2} 筆電壓的平均值: {avg2:.1f}V")

# 只需要「平均值」，不需要「數量」時，將不需要的 count 位置，用底線_來接收
avg3,_= calculate_average_voltage(10,20,30)
print(f"電壓的平均值: {avg3:.1f}V")

# 或是可以直接呼叫完函式後，用 [0] 拿取第 0 個位置的資料 (也就是平均值)
avg3= calculate_average_voltage(10,20,30)[0]
print(f"電壓的平均值: {avg3:.1f}V")

# 如果函式回傳多個數值但只需要部分數值時，將不需要的變數位置，用 *_ 來接收
# 一個星號 *_ 代表：「剩下的不管有幾百個，通通打包丟進這個廢棄垃圾桶裡！」
avg4,*_= calculate_average_voltage(10,30,50)
print(f"電壓的平均值: {avg4:.1f}V")

# 取函式回傳的第一個數值 與 最後一個數值
# 抓第一個用 first，抓最後一個用 last，中間剩下的通通丟棄到 *_
first, *_, last= calculate_average_voltage(20,40,60)
print(f"{last} 筆電壓的平均值: {first:.1f}V")

```

```python
# def: 函式
print("--------(def: 函式的定義與呼叫)--------")

#=================================
#      基本語法:
#
#  def 函式名稱(參數名稱):
#   函式內部的程式碼
#=================================
def say_hello():
  print("你好! 歡迎來到 Python 的世界!")
say_hello()


# print 只負責把文字顯示在螢幕上給人看，印出來但是不會記錄在電腦裡 (不會保存資料)
name=input("請輸入名稱: ")
def say_hello(name):
  print(f"你好! {name} 歡迎來到 Python 的世界!")
say_hello(name)

# return 則是把運算結果「交還」給程式，讓外面的變數可以抓取繼續使用
def make_greeting(name):
  return f"你好! {name} 歡迎來到 Python 的世界!"
result=make_greeting(name)
print(result)
new_message=result + "祝你有一個美好的一天!"
print(new_message)


# 計算客廳面積
length1 = 5
width1 = 4
area1 = length1 * width1
print("客廳的面積是:", area1)

# 計算臥室面積
length2 = 3
width2 = 3
area2 = length2 * width2
print("臥室的面積是:", area2)

# 使用 def函式 打包
def calculate_area(length, width):
  area=length*width
  return area
print("客廳的面積是:", calculate_area(5,4))
print("臥室的面積是:", calculate_area(3,3))
print(f"客廳(5X4)加上臥室(3X3)的總面積是: {calculate_area(5,4)+calculate_area(3,3)}")

```

```python
# def: 函式

# 參數的預設資料
print("--------(def: 參數的預設資料)--------")
def power(base,exp):
  print(base**exp)
power(3,2)
#power(4)    #輸出錯誤: 缺少了"exp"的參數值

def power(base,exp=0):  #預設exp的參數值為 0
  print(base**exp)
power(3,2)
power(4)    #正確輸出: exp預設參數值為 0 的值(power(4,0))

# 補充題目練習: 不使用 內建運算"**" 表示次方
def power(base,exp):
  result=1
  for i in range(exp):
    result=result*base
  return result
print(power(3,2))


# 使用參數名稱對應
print("--------(def: 使用參數名稱對應)--------")
def divide(n1,n2):
  print(n1/n2)

divide(2,4)
divide(n2=2,n1=4)


# *args: 不定數量參數
print("--------(def: 不定數量參數)--------")
def avg(*num):
  count=len(num)
  total=0
  for n in num:
    total=total+n
  result= total/count
  print(result)

avg(3,4)
avg(3,5,10)
avg(1,4,-1,8)
```
