```python
#==========================
# 1. 取得現在的日期與時間
#==========================
from datetime import datetime, date

# 取得現在完整的日期與時間 (年-月-日 時:分:秒.微秒)
now= datetime.now()
print(f"現在時間: {now}")

# 只取得今天的日期
today= date.today()
print(f"今天日期: {today}")

# 也可以單獨取出年份、月份或日期
print(f"今年是 {now.year} 年，現在是 {now.month} 月")


#==========================
# 2. 字串與時間的互相轉換
#==========================

# 實務上，常需要把時間變成特定格式的字串印出來，或者把使用者輸入的字串（例如: "2026-08-05"）轉換成程式能辨識的時間物件
# 會用到兩個長得很像的方法：.strftime() 與 .strptime()

# .strftime()  (String Format Time)： 把時間物件變成字串
#  使用情境：當需要把時間印到畫面上、當作檔案名稱存檔、或是顯示在視窗介面 (GUI) 上時

# .strptime()  (String Parse Time)： 把字串解析成時間物件
#  使用情境：當需要讀取使用者的輸入、讀取 CSV 檔案裡的日期資料，而且後續需要對這些時間進行加減運算 (timedelta) 或比較大小時

from datetime import datetime
now = datetime.now()

# 【時間 轉換為 字串】
# %Y:四位數年份， %m:兩位數月份， %d:日期， %H:24小時制， %M:分鐘， %S:秒
time_str= now.strftime("%Y/%m/%d %H:%M:%S")
print(f"自訂格式輸出: {time_str}")

# 【字串轉換為時間】
user_input= "2026-12-31"
# 告訴 Python 這個字串的格式，就能轉換
parsed_date= datetime.strptime(user_input, "%Y-%m-%d")
print(f"轉換後的物件型態: {type(parsed_date)}")


#===================
# 3. 日期的加減法
#===================
from datetime import datetime, timedelta
now= datetime.now()

# 建立一個「時間差」的物件，設定為 7 天
seven_days= timedelta(days=7)

# 計算未來的時間 (現在時間 + 7天)
future_date= now + seven_days
print(f"七天後的時間是: {future_date.strftime('%Y-%m-%d')}")

# 計算過去的時間 (現在時間 - 3小時)
three_hours_ago= now - timedelta(hours=3)
print(f"三小時前的時間是: {three_hours_ago.strftime('%H:%M:%S')}")


#========================
# 4.計算兩個時間的差距
#========================
from datetime import datetime

# 設定一個目標時間 (例如: 2026年12月31日 23點59分59秒)
target_time= datetime(2026, 12, 31, 23, 59, 59)
now= datetime.now()

# 兩個時間相減
time_left= target_time - now
print(f"距離跨年還有: {time_left.days} 天")
print(f"總共還剩下: {time_left.total_seconds()} 秒")          #也可以換算成總秒數




#=========================================================
# datetime + timedelta = datetime ➔ 可以用 .strftime()
# datetime - timedelta = datetime ➔ 可以用 .strftime()
# datetime - datetime = timedelta ➔ 不能用 .strftime()，
# 只能讀取 .days 或 .total_seconds()
#=========================================================
```
