```python
print("--------(Pandas: Series 單維度資料)--------")

# 載入 Pandas 模組
import pandas as pd

# 資料索引

data= pd.Series([5, 4, -2, 3, 7])
print(f"預設索引\n{data}\n")

data= pd.Series([5, 4, -2, 3, 7],index= ["a","b","c","d","e"])
print(f"自訂索引\n{data}\n")


# 觀察資料
print(f"資料型態: {data.dtype}")
print(f"資料數量: {data.size}")
print(f"資料索引: {data.index}")


# 取得資料: 根據順序、根據索引

# 根據順序取得第三筆資料
print(f"\n取得第三筆資料: {data.iloc[2]}")           #輸出: -2

# 根據索引 "e" 取得對應的資料
print(f"根據索引 'e' 取得對應的資料: {data['e']}")    #輸出: 7


# 數字運算: 基本、統計、順序
print(f"\n最大值: {data.max()}")
print(f"總和: {data.sum()}")
print(f"標準差: {data.std()}")
print(f"中位數: {data.median()}")
print(f"\n所有數值放大兩倍\n{data*2}\n")
print(f"最大的三個數值\n{data.nlargest(3)}\n")
print(f"最小的兩個數值\n{data.nsmallest(2)}\n")


# 字串運算: 基本、串接、搜尋、取代
data= pd.Series(["您好", "Python", "Pandas"])

print(f"所有字串變小寫\n{data.str.lower()}\n")
print(f"所有字串變大寫\n{data.str.upper()}\n")
print(f"算出每一個字串的長度\n{data.str.len()}\n")
print(f"把所有字串用逗號串在一起\n{data.str.cat(sep=',')}\n")
print(f"判斷字串中是否包含 'P'\n{data.str.contains('P')}\n")
print(f"用 'Hello' 取代字串中的 '您好'\n{data.str.replace('您好','Hello')}")



#========================
# Series 單維度資料
#========================

# 建立 Series
# import pandas as pd
# 以列表資料為基底，建立 Series
# pd.Series(資料列表)

#-------------------------------------------------------------

# 資料索引: 資料的獨立編號，像是試算表中最左邊的編號

# 內建索引
# 載入 Pandas 模組
# 以列表資料為基底，建立 Series
# pd.Series(資料列表)

# 自訂索引
# 載入 Pandas 模組
# 以列表資料為基底，建立 Series
# pd.Series(資料列表, index=索引列表)

#-------------------------------------------------------------

# 觀察資料

# import pandas as pd
# data= pd.Series(資料列表)

# .dtype: 觀察資料「型態」
# print(data.dtype)

# .size: 觀察資料「數量」
# print(data.size)

# .index: 觀察資料「索引」
# print(data.index)

#-------------------------------------------------------------

# 取得資料

# 根據資料的順序取值
# import pandas as pd
# data= pd.Series(資料列表)
# 取得資料 data.iloc[順序]
# print(data.iloc[1])      #取得第二個資料

# 根據索引取值
# import pandas as pd
# data= pd.Series(資料列表, index=索引列表)
# 取得資料 data[索引]
# print(data[索引])

#-------------------------------------------------------------

# 數學運算

# 數學、統計相關
# import pandas as pd
# data= pd.Series([3, 10, 20, 5, -12])

# 各種數學、統計運算

# 算出加法的總和、找出最大值、算乘法的總和
# print(data.sum(), data.max(), data.prod())

# 算出平均數、中位數、標準差
# print(data.mean(), data.median(), data.std())

# 取得前 3 最大的數值、取得前 2 最小的數值
# print(data.nlargest(3), data.nsmallest(2))

#-------------------------------------------------------------

# 字串運算

# 字串相關的操作
# import pandas as pd
# data= pd.Series(["您好", "Python", "Pandas"])

# 各種字串操作，都定義在 str 底下

# 把所有的字串變成小寫、把所有的字串變成大寫、得到每個字串的長度
# print(data.str.lower(), data.str.upper(), data.str.len())

# 把所有字串用 「逗號 ,」 串在一起、判斷每個字串中是否包含 "P"
# print(data.str.cat(sep=','), data.str.contains('P'))

# 將每一個字串中的 "您好" 使用 "Hello" 取代
# print(data.str.replace("您好","Hello"))
```
