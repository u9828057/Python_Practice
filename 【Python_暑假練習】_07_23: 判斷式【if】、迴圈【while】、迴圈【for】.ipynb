```python
# 判斷式
if True:
  print("True 執行")
else:
  print("False 執行")

x=input("請輸入數字: ")      #取得字串形式的使用者輸入
x=int(x)              #將字串型態轉換為數字型態
if x>100:
  print("大於100")
else:
  print("小於100")

if x>200:
  print("大於200")
elif x>100:
  print("大於100，小於等於200")
else:
  print("小於100")


# 四則運算
n1=int(input("請輸入數字一: "))
n2=int(input("請輸入數字二: "))
print(n1+n2)

op=input("請輸入運算: +, -, *, /")
if op=="+":
  print(n1+n2)
elif op=="-":
  print(n1-n2)
elif op=="*":
  print(n1*n2)
elif op=="/":
  if n2==0:               #巢狀判斷：在除法運算之前，先進行邊界防呆（因為在數學與電腦運算中，除數絕對不能為0）
    print("除數不可為零")
  else:
    print(n1/n2)
else:
  print("不支援運算")
```

```python
# while 迴圈
print("--------(while 迴圈)--------")
n=1
while n<=3:
  print(n)
  n+=1

# 1+2+3+...+10
n=1
total=0       #紀錄累加的結果
while n<=10:
  total=total+n
  n+=1
print(total)

# for 迴圈
print("--------(for 迴圈)--------")
print("--------(for x in [3,5,1])--------")
for x in [3,5,1]:
  print(x)

print(f"\n--------(for x in \"Hello\")--------")
for x in "Hello":
  print(x)

print("--------(for x in range(5))--------")
for x in range(5):
  print(x)
#range(5)代表從0開始，產生5個數字：0,1,2,3,4(不包含5)

print("--------(for x in range(5,10))--------")
for x in range(5,10):
  print(x)
#range(起始,結束)規則：包含開頭，不包含結尾(印出5,6,7,8,9)

print("--------(for x in range(1,11), Total+=x)--------")
total=0
for x in range(1,11):
  total=total+x
print(total)
#因為要累加到10，所以結束值必須設定為11
```

```python
# break 的簡易範例
print("--------(break 的簡易範例)--------")
n=0
while n<5:
  if n==3:
    break
  n+=1
  print(n)             #印出退出迴圈中的 n值
print(f"最後的 n: {n}")      #印出迴圈結束後的 n值


# continue 的簡易示範
print("--------(continue 的簡易示範)--------")
n=0
for x in [0,1,2,3]:
  if x%2==0:     #x是偶數
    continue
  print(x)
  x+=1
print(f"最後的 x: {x}")

#break   = 「直接退場，徹底打破迴圈」   (比喻: 徹底結束迴圈，不再執行歌曲)
#continue = 「略過當次，煞車跳回頂端」   (比喻: 繼續運轉迴圈，只是跳過一首歌)


# else 的簡易範例
print("--------(else 的簡易範例)--------")
total=0
for n in range(11):
  total+=n
else:
  print(total)          #印出 0+1+2+...+10的結果


# 綜合範例: 找出整數平方根
# 輸入9, 得到 3
# 輸入11, 得到 【沒有】整數的平方根
n=input("輸入一個正整數: ")
n=int(n)              #輸入 轉換為數字
for i in range(n+1):       #i從 0～n
  if i*i==n:
    print("整數平方根", i)
    break           #用 break 強制結束迴圈時，不會執行else區塊
else:
  print("沒有整數平方根")



#Python 創始人的命名冷知識
#Python界的冷知識：連Python的創始人Guido van Rossum 後來都公開承認，當年把這個語法命名為 else 是一個糟糕的決定！
#如果能夠重新設計 Python，他會把 for ... else 的那個 else 改名為 no_break。
#只要記住 no_break 這個概念，邏輯就絕對不會錯：
#沒遇到  break ➡️ 迴圈從頭到尾平安走完 ➡️ 執行 else。
#中途遇到 break ➡️ 迴圈被半路攔截殺死  ➡️ 直接跳過，不執行 else。
```
