```python
import numpy as np
import matplotlib.pyplot as plt
print("--------(Matplotlib: 簡介與快速開始)--------")

#=====================================
# 1. 準備繪圖數據 (結合 NumPy)
#=====================================
print("\n--------(準備實驗數據)--------")

# 繪圖專用的 NumPy 技巧: np.linspace(起點， 終點， 數量)

# 範例: 會在 0 到 10 之間，均勻產生 100 個點 (非常適合用來繪製平滑的曲線)
time = np.linspace(0, 10, 100)

# 假設這是一個振幅為 5V 的正弦波電壓訊號: V(t)= 5*sin(t)
voltage = 5*np.sin(time)

print(f"\n【時間陣列形狀】: {time.shape}")
print(f"【電壓陣列形狀】: {voltage.shape}")
print(f"「確保 X 軸與 Y 軸的資料數量完全一致，是畫圖的鐵則」")


#==============================
# 2. 開始繪圖 (Plotting)
#==============================
print("\n--------(產生波形圖)--------\n")

# 步驟 A: 繪製基本折線
# 語法邏輯: plt.plot(X軸資料， Y軸資料)
plt.plot(time, voltage)


# 步驟 B: 加上圖表元素
plt.title("Voltage / Time (Sine Wave)")   #圖表標題
plt.xlabel("Time (s)")             #X 軸標籤
plt.ylabel("Voltage (V)")            #Y 軸標籤
plt.grid(True)                  #打開網格線


# 步驟 C: 顯示繪圖
plt.show()
```
