```python
import matplotlib.pyplot as plt
plt.rc("font", family="Noto Sans CJK TC")
print("--------(Matplotlib: 折線圖)--------\n")

# 實際操作: 熟悉折線圖的操作

"""
  一組資料點
  (1, 2),(2, 4),(3, 6)
"""

plt.plot([1, 2, 3],[2, 4, 6])
plt.show()

"""
  兩組資料點
   第一組 (1, 2),(2, 4),(3, 6)
   第二組 (1, 1),(2, 2),(3, 3)
"""

# 設定每一組資料與對應的標籤
plt.plot([1, 2, 3],[[2, 1],[4, 2],[6, 3]], label=["第一組標籤","第二組標籤"])

plt.legend()  #產生圖例
plt.xlabel("X 軸的文字說明")
plt.ylabel("Y 軸的文字說明")
plt.show()

#-------------------------------------------------------------

# 從 CSV 格式的檔案載入資料，並繪製折線圖
import csv
print("\n--------(從 CSV 格式載入資料，並繪製折線圖)--------\n")

# 1. 準備 CSV 資料數據
with open("data.csv", mode="w", encoding="utf-8", newline="") as file:
  w = csv.writer(file)
  w.writerow(["年度","小王","小美"])
  w.writerow([2010, 40000, 30000])
  w.writerow([2011, 42000, 60000])
  w.writerow([2012, 45000, 50000])

# 2. 讀取資料並整理成三個獨立的陣列
with open("data.csv", mode="r", encoding="utf-8", newline="") as file:
  r = csv.DictReader(file)
  header = r.fieldnames  #讀取標題
  print(f"標頭: {header}")

  x=[]     #存放 X 軸: 年度
  y_wang=[]   #存放第一條線: 小王
  y_mei=[]   #存放第二條線: 小美

  for row in r:
    print(f"年度的資料: {row["年度"]}")
    x.append(int(row["年度"]))
    y_wang.append(int(row["小王"]))
    y_mei.append(int(row["小美"]))

# 3. 繪製圖表
# 呼叫兩次 plot，分別繪製兩個人的折線，並動態抓取 header 作為圖例名稱
plt.plot(x, y_wang, label=header[1])
plt.plot(x, y_mei, label=header[2])

# 加上圖表元素
plt.title("年度薪資比較表")
plt.xlabel(header[0])
plt.ylabel("薪水")
plt.grid(True)

plt.legend();  #呼叫出圖例
plt.show()










#=======================================
# Matplotlib 折線圖: 資料點的呈現
# 基礎的使用
#=======================================

# 使用 plot 方法
# 根據資料點畫出折線圖

# 資料點 (x, y) 範例如下:
# (2, 4),(3, 6),(4, 8)
# 程式碼:
plt.plot([2, 3, 4], [4, 6, 8])


# 兩組資料點
# 在同一張圖上畫出兩個折線圖

# 資料點 (x, y) 範例如下:
# 第一組 (2, 4),(3, 6),(4, 8)
# 第二組 (2, 2),(3, 3),(4, 4)
# 程式碼:
plt.plot([2, 3, 4],[[4, 2],[6, 3],[8, 4]])


#===================================
# 圖例與軸線標題: 圖表中的中文
# 設定支援中文的字型
#===================================

# 使用 rc 方法，設定執行環境:
# plt.rc("font", family="字型的名稱")
# 程式碼:
plt.rc("font", family="NoTo Sans CJK TC")

#-------------------------------------------------------------

# §補充資料: Colab 畫圖中文顯示起手式
import matplotlib.font_manager as fm

# 下載微軟正黑體 (或是 NoTo 繁體中文)

# 1. 透過指令下載 Google 的 Noto Sans 繁體中文字型 (這行指令適用於 Colab 或 Jupyter 筆記本)
!wget -q -O NotoSansTC-Regular.otf https://github.com/googlefonts/noto-cjk/raw/main/Sans/OTF/TraditionalChinese/NotoSansCJKtc-Regular.otf
#語法解釋: 是一個 Linux 系統指令，意思是：「去這個網址，把一個叫做 Noto 的中文字體檔案下載下來」

# 2. 將下載好的字體加入 Matplotlib 的字體庫中
fm.fontManager.addfont("NotoSansTC-Regular.otf")
#語法解釋: 把剛剛下載下來的字體檔案，正式註冊到 Matplotlib 的字體庫裡面

# 3. 設定 Matplotlib 使用剛剛載入的中文字體
plt.rc("font", family="Noto Sans CJK TC")
#語法解釋: 告訴 Matplotlib 「從現在開始，把所有的預設字體都換成這個新下載的中文字體」

# 4. 解決有時候座標軸負號 (-) 顯示成方塊的問題
plt.rcParams["axes.unicode_minus"] = False
#語法解釋: 是一個常見的 Bug 修復，當換了中文字體後，座標軸上的「負號（-）」有時候會變成奇怪的方塊，這行程式碼就是專門用來解決這個問題的


#==================================
# 標籤與圖例
# 使用 label 參數設定標籤
# 使用 legend 方法，產生圖例
#==================================

# 資料點 (x, y) 範例如下:
# 第一組 (2, 4),(3, 6),(4, 8)
# 第二組 (2, 2),(3, 3),(4, 4)
# 程式碼:
plt.plot(
    [2, 3, 4],
    [[4, 2],[6, 3],[8, 4]],
    label=["第一組","第二組"]
)

# 設定標籤後，呼叫 legend 產生圖例
plt.legend();


#====================================
# 軸線的標題
# 使用 xlabel 與 ylabel 方法
# 設定軸線標題
#====================================

# 程式碼:
plt.plot([2, 3, 4],[4, 6, 8])
plt.xlabel("X 軸的標題")
plt.ylabel("Y 軸的標題")


# 演練目標
# 熟悉折線圖的操作
# 繪製一組資料點的折線圖
# 繪製兩組資料點的折線圖，並且加上適當的標題
# 載入 CSV 格式的檔案資料，繪製折線圖並加上適當的標題
```

```python
#==================
# 實驗模擬情境
#==================
import numpy as np
import matplotlib.pyplot as plt
print("--------(Matplotlib: 折線圖)--------\n")

# 1. 準備數據
time = np.linspace(0, 10, 50)  #取 50 個取樣點
v_in = 5*np.sin(time)       #輸入訊號 (Sine)
v_out = 5*np.cos(time)      #輸出訊號 (Cosine)

#-------------------------------------------------------------

# 2. 繪製折線圖
plt.figure(figsize=(8, 5))  #設定整張圖表的長寬比例 (8 x 5)

# 繪製第一條線 (紅色、虛線、圓點標記)
# 並加上 label="Input (Sine)" 準備給圖例使用
plt.plot(time, v_in, "r--o", label="Input (Sine)")

# 繪製第二條線 (藍色、實線、無標記，增加線條寬度)
plt.plot(time, v_out, color="blue", linestyle="-", \
         linewidth=2, label="Output (Cosine)")

#-------------------------------------------------------------

# 3. 完善圖表資訊
plt.title("電壓比較表: 輸入/輸出", fontsize=14)  #fontsize 可以調整字體大小
plt.xlabel("Time (s)")
plt.ylabel("Voltage (V)")
plt.axhline(y=0, color="black",linestyle=":",linewidth=1.5,label="0V 基準線")
plt.grid(True)
plt.legend();   #顯示圖例 (自動抓取前面設定的 label)


# §補充資料: 可以使用 plt.axhline() 或 plt.axvline() 在圖表中加入一條貫穿全圖的水平或垂直輔助線
# .axhline() 代表 Axis Horizontal Line
# .axvline() 代表 Axis Vertical Line
# 可以將 y=0 設定在任何的電壓準位，這在觀察放大器是否有截波（Clipping）失真時相當好用

#-------------------------------------------------------------

# 4. 儲存與顯示圖片

# 將畫好的圖存成高畫質 PNG 檔
# dpi=300 代表高解析度，bbox_inches="tight" 可以避免存檔時邊緣的標籤被裁切掉
plt.savefig("電壓比較表 - 輸入_輸出.png", dpi=300, bbox_inches="tight")
print(f"圖片已經成功儲存為 '電壓比較表 - 輸入_輸出.png'")

# 最後才顯示圖片
plt.show()
```
