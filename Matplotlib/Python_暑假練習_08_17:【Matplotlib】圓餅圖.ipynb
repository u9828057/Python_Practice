```python
import matplotlib.pyplot as plt
print("--------(Matplotlib: 圓餅圖)--------")

# 實際操作: 熟悉圓餅圖的操作

# 繪製一組資料點的圓餅圖，加上適當的百分比標示
plt.pie(
    [10, 15, 25],
    labels=["10", "15", "25"],
    labeldistance=0.5
)

# 顯示標籤與圖片
plt.legend()  #呼叫圖例
plt.show()

#-------------------------------------------------------------

# 繪製一組資料點的圓餅圖，加上適當的百分比標示
x = [10, 15, 25]
total = sum(x)  #算列表中的數值總和 sum(列表)

#-----------------------------------------------------------
# §寫法一: 手動土法煉鋼 (逐一寫出 x[0], x[1], x[2])
#-----------------------------------------------------------

plt.pie(
    x,
labels=[f"{100*x[0]/total}%",f"{100*x[1]/total}%",f"{100*x[2]/total}%"]
    ,labeldistance=0.5
)

# 顯示標籤與圖片
plt.title("寫法一: 手動寫出索引")
plt.legend();
plt.show()

#--------------------------------------
# §寫法二: for 迴圈 (列表推導式)
#--------------------------------------

# 一行動態計算所有標籤
labels = [f"{100*n / total}%" for n in x]
print(f"查看 labels: {labels}\n")

plt.pie(
    x,
    labels=labels,
    labeldistance=0.5
)

# 顯示標籤與圖片
plt.title(f"寫法二: 列表推導式")
plt.legend();
plt.show()

#-------------------------------------------------------------

# 載入 CSV 格式檔案的資料，繪製適當的圓餅圖加標示
import csv
print("\n--------(從 CSV 格式載入資料，並繪製圓餅圖)--------\n")

# 1. 準備 CSV 資料數據
with open("pie-chart-data.csv", mode="w", encoding="utf-8", newline=""
    ) as file:
    w = csv.writer(file)
    w.writerow(["瀏覽器", "市佔率"])
    w.writerow(["Chrome", 60])
    w.writerow(["Safari", 20])
    w.writerow(["Edge", 10])
    w.writerow(["Firefox", 10])

# 2. 讀取資料
with open("pie-chart-data.csv", mode="r", encoding="utf-8", newline=""
    ) as file:
    r = csv.DictReader(file)
    header = r.fieldnames  #讀取標題
    print(f"標頭: {header}")

    labels = []  #存放 "瀏覽器"
    x = []  #存放 "市佔率"

    for row in r:
      print(f"瀏覽器: {row['瀏覽器']}")
      print(f"市佔率: {row['市佔率']}")
      labels.append(row['瀏覽器'])
      x.append(int(row['市佔率']))

# 3. 繪製圖表
plt.figure(figsize=(6, 6))  #建立 6x6 畫布，讓圖片更清楚

plt.pie(
    x,
    labels=labels,
    labeldistance=1.1,
    # autopct 用來設定數值格式:
    # 1. %開頭表示格式化
    # 2. .1f 表示浮點數顯示到小數點後 1 位
    # 3. %% 用來在畫面上顯示出百分比符號
    autopct="%1.1f%%"
)

# 4. 顯示圖片
plt.title("瀏覽器的市場佔有率分佈")
plt.show()








#=======================================
# Matplotlib 圓餅圖: 資料點的呈現
# 基礎的使用
#=======================================

# 使用 pie 方法
# 根據一個維度的資料點畫出圓餅圖

# 資料點範例如下:
# 20, 30, 40
# 程式碼:
# plt.pie([20, 30, 40])


#===================================
# 標籤、圖例與標題: 標題與圖例
# 使用 labels 參數設定標籤
# 使用 legend 方法，產生圖例
#===================================

# 資料點 x 範例如下:
# 20, 30, 40
# 程式碼:
"""
plt.pie(
    [20, 30, 40],
    labels=["第一塊","第二塊","第三塊"]
)
"""
# 設定標籤後，呼叫 legend 產生圖例
# plt.legend()


#===========================================
# 標籤的位置
# 使用 labeldistance 參數設定標籤位置
#===========================================

# 0 代表圓心，1 代表圓周
# 程式碼:
_="""
plt.pie(
    [20, 30, 40],
    labels=["第一塊","第二塊","第三塊"],
    labeldistance=0.5
)
"""


#===================================
# 圖表的標題
# 使用 title 方法設定圖表標題
#===================================

# 程式碼:
# plt.pie([20, 30, 40])
# plt.title("這是一個圓餅圖")

# 演練目標
# 熟悉圓餅圖的操作
# 繪製一組資料點的圓餅圖，並且加上適當的百分比標示
# 載入 CSV 格式的檔案資料，繪製圓餅圖並加上適當的標示
```

```python
import matplotlib.pyplot as plt
print("--------(Matplotlib: 圓餅圖)--------\n")

# 1. 準備數據
# 假設這是一天 24 小時的時間分配
activities = ["工程數學與實驗", "整理筆記進度", "睡眠與休息", "休閒與其他"]
hours = [7, 4, 7, 6]

# 2. 設定「凸顯」效果 (Explode)
# 有 4 個項目，所以需要一個長度為 4 的 Tuple
# 將第二個項目 (整理筆記進度) 設定為 0.1，單獨凸顯出來
explode_setting = (0, 0.1, 0, 0)

# 3. 繪製圖片
plt.figure(figsize=(6, 6))  #讓畫布呈現 6x6 的正方形

plt.pie(
    hours,
    labels=activities,
    autopct="%1.1f%%",
    explode=explode_setting,
    startangle=90,
    shadow=True,  #shadow=True 可以增加一點陰影，增加立體感
    # wedgeprops 用來製作甜甜圈圖:
    # 1. width=0.4 表示保留外圈 40% 的厚度
    # 2. edgecolor="white" 會加上白色邊框讓區塊更分明
    wedgeprops={"width": 0.4, "edgecolor": "white"}
)

# 4. 完善圖表資訊
plt.title("每日時間分配比例圖", fontsize=16)

#【重要】: 確保圓餅圖畫出來是正圓形，不會因為視窗比例拉伸變成橢圓形
plt.axis("equal")

# 5. 儲存並顯示圖片
plt.savefig("每日時間分配比例圖.png", dpi=300, bbox_inches="tight")
print("圖片儲存完畢!")

plt.show()


#=======================================================
# §補充技巧
# 為了讓圓餅圖看起來更專業，通常會搭配幾個進階參數
#=======================================================

# x（必要）：數值陣列（例如各個項目的花費金額或時間）

# labels：每個扇形區塊的名稱標籤

# explode：凸顯特定區塊（爆炸效果），可以設定一個陣列，數值代表該區塊距離圓心的距離，
# 0 代表在圓心，0.1 代表拉出 10% 的半徑距離

# autopct：自動計算百分比（Auto Percentage），是一個格式化字串，
# 最常見的寫法是 '%1.1f%%'，意思是「顯示小數點後 1 位數，並在最後加上 % 符號」

# startangle：起始角度，Matplotlib 預設是從 3 點鐘方向（0 度）逆時針開始畫，
# 在工程簡報中，通常會設定 startangle=90，讓第一塊資料從 12 點鐘方向開始，視覺上更為直覺
```
