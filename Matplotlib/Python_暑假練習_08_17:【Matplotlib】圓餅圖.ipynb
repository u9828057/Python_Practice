```python
import matplotlib.pyplot as plt
print("--------(Matplotlib: 圓餅圖)--------")

# 實際操作: 熟悉圓餅圖的操作

# 繪製一組資料點的圓餅圖，加上適當的百分比標示
plt.pie(
    [10, 15, 25],
    labels=["10", "15", "25"],
    labeldistance=0.5
)

# 顯示標籤與圖片
plt.legend()  #呼叫圖例
plt.show()

#-------------------------------------------------------------

# 繪製一組資料點的圓餅圖，加上適當的百分比標示
x = [10, 15, 25]
total = sum(x)  #算列表中的數值總和 sum(列表)

#-----------------------------------------------------------
# §寫法一: 手動土法煉鋼 (逐一寫出 x[0], x[1], x[2])
#-----------------------------------------------------------

plt.pie(
    x,
labels=[f"{100*x[0]/total}%",f"{100*x[1]/total}%",f"{100*x[2]/total}%"]
    ,labeldistance=0.5
)

# 顯示標籤與圖片
plt.title("寫法一: 手動寫出索引")
plt.legend();
plt.show()

#--------------------------------------
# §寫法二: for 迴圈 (列表推導式)
#--------------------------------------

# 一行動態計算所有標籤
labels = [f"{100*n / total}%" for n in x]
print(f"查看 labels: {labels}\n")

plt.pie(
    x,
    labels=labels,
    labeldistance=0.5
)

# 顯示標籤與圖片
plt.title(f"寫法二: 列表推導式")
plt.legend();
plt.show()

#-------------------------------------------------------------

# 載入 CSV 格式檔案的資料，繪製適當的圓餅圖加標示
import csv
print("\n--------(從 CSV 格式載入資料，並繪製圓餅圖)--------\n")

# 1. 準備 CSV 資料數據
with open("pie-chart-data.csv", mode="w", encoding="utf-8", newline=""
    ) as file:
    w = csv.writer(file)
    w.writerow(["瀏覽器", "市佔率"])
    w.writerow(["Chrome", 60])
    w.writerow(["Safari", 20])
    w.writerow(["Edge", 10])
    w.writerow(["Firefox", 10])

# 2. 讀取資料
with open("pie-chart-data.csv", mode="r", encoding="utf-8", newline=""
    ) as file:
    r = csv.DictReader(file)
    header = r.fieldnames  #讀取標題
    print(f"標頭: {header}")

    labels = []  #存放 "瀏覽器"
    x = []  #存放 "市佔率"

    for row in r:
      print(f"瀏覽器: {row['瀏覽器']}")
      print(f"市佔率: {row['市佔率']}")
      labels.append(row['瀏覽器'])
      x.append(int(row['市佔率']))

# 3. 繪製圖表
plt.figure(figsize=(6, 6))  #建立 6x6 畫布，讓圖片更清楚

plt.pie(
    x,
    labels=labels,
    labeldistance=1.1,
    # autopct 用來設定數值格式:
    # 1. %開頭表示格式化
    # 2. .1f 表示浮點數顯示到小數點後 1 位
    # 3. %% 用來在畫面上顯示出百分比符號
    autopct="%1.1f%%"
)

# 4. 顯示圖片
plt.title("瀏覽器的市場佔有率分佈")
plt.show()








#=======================================
# Matplotlib 圓餅圖: 資料點的呈現
# 基礎的使用
#=======================================

# 使用 pie 方法
# 根據一個維度的資料點畫出圓餅圖

# 資料點範例如下:
# 20, 30, 40
# 程式碼:
# plt.pie([20, 30, 40])


#===================================
# 標籤、圖例與標題: 標題與圖例
# 使用 labels 參數設定標籤
# 使用 legend 方法，產生圖例
#===================================

# 資料點 x 範例如下:
# 20, 30, 40
# 程式碼:
"""
plt.pie(
    [20, 30, 40],
    labels=["第一塊","第二塊","第三塊"]
)
"""
# 設定標籤後，呼叫 legend 產生圖例
# plt.legend()


#===========================================
# 標籤的位置
# 使用 labeldistance 參數設定標籤位置
#===========================================

# 0 代表圓心，1 代表圓周
# 程式碼:
_="""
plt.pie(
    [20, 30, 40],
    labels=["第一塊","第二塊","第三塊"],
    labeldistance=0.5
)
"""


#===================================
# 圖表的標題
# 使用 title 方法設定圖表標題
#===================================

# 程式碼:
# plt.pie([20, 30, 40])
# plt.title("這是一個圓餅圖")

# 演練目標
# 熟悉圓餅圖的操作
# 繪製一組資料點的圓餅圖，並且加上適當的百分比標示
# 載入 CSV 格式的檔案資料，繪製圓餅圖並加上適當的標示
```
