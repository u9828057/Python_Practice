```python
import matplotlib.pyplot as plt
print("--------(Matplotlib: 長條圖)--------\n")

# 1. 準備數據
subjects = ["電子學(二)", "電子學實驗(二)", "電磁學(一)", "工程數學"]
scores = [85, 92, 78, 88]


# 2. 繪製長條圖
plt.figure(figsize=(8, 5))  #建立 8x5 畫布，讓圖片更清楚

# 使用 bar 繪圖，並帶入進階參數
# width=0.6: 讓柱子變瘦一點，視覺上更舒適
# color: 設定自訂的顏色
# edgecolor="black": 加上黑色邊框
bars = plt.bar(subjects, scores,
        width=0.6, color="#4A90E2",
        edgecolor="black"
        )


# 3. 完善圖表資訊
plt.title("學期期中考成績比較圖", fontsize=15, color="#FF7744")
plt.xlabel("課程名稱", fontsize=12, color="#AAAAAA")
plt.ylabel("分數", fontsize=12, color="#AAAAAA")

# §實用技巧: 設定 Y 軸的顯示範圍 (0 到 100)
# 可以避免 Matplotlib 自動縮放，導致 78 分看起來像是 0 分的錯覺
plt.ylim(0, 100)
plt.grid(axis="y", linestyle="--", alpha=0.7)  #只留橫向網格線，畫面更乾淨


# 4. 【進階技巧】:在柱子上直接標示數值
# 使用 plt.bar_label 可以自動把數值寫在長條圖的頂端
plt.bar_label(bars, padding=3, fontsize=11)


# 5. 儲存並顯示圖片
plt.savefig("學期期中考成績比較圖.png", dpi=300, bbox_inches="tight")
plt.show()










# §補充資料

#----------------------------------------------------
# 1. plt.bar_label(padding=數值): 控制標籤間距
#----------------------------------------------------
# [用途]: 調整 「數值文字」 與 「長條圖頂端」 的距離，保持視覺上的舒適度，讓圖表更乾淨與專業
# [參數]: padding (單位為: 像素/點)
# [場景]: 若設定 padding=0，數字會剛好「貼死」在柱子最上緣，視覺上較為擁擠，通常設定為 3～5 來將文字往上推

# 範例: 在長條圖上標示數值，並將文字往上推 3 個像素
# plt.bar_label(bars, padding=3, fontsize=11)

#---------------------------------------------------
# 2. plt.grid(linestyle="樣式"): 控制線條外觀
#---------------------------------------------------
# [用途]: 改變繪圖線或網格線的樣式，工程製圖中，通常將輔助網格線改為「虛線」，避免干擾資料主角(柱子或折線)的呈現
# [參數]: linestyle (常簡寫為 ls)，傳入特定的字串符號

# 常用的線條格式字典:
# "-" : 實線（Solid line，預設值）
# "--" : 虛線（Dashed line，推薦用於網格線）
# ":" : 點線（Dotted line，由密集的點組成）
# "-." : 點虛線（Dash-dot line，一長一短交錯，常用於標示對稱軸或中心線）

# 範例: 開啟 Y 軸的網格線，並設定為半透明的虛線
# plt.grid(axis="y", linestyle="--", alpha=0.7)
```
