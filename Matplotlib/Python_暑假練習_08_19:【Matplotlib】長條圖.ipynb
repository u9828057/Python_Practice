```python
import matplotlib.pyplot as plt
print("--------(Matplotlib: 長條圖)--------\n")

# 實際操作: 熟悉長條圖的操作

"""
  X 軸為數字
   x = [3, 4, 1]
   height = [8, 5, 2]
"""
plt.bar([3, 4, 1], [8, 5, 2])
plt.show()

"""
  X 軸為字串，調整寬度與顏色，加上標示
  x = ["B", "A", "C"]
  height = [8, 10, 16]
"""
plt.bar(["B", "A", "C"], [8, 10, 16], width=0.5, color="red")
plt.xlabel("測試的 X 軸標籤")
plt.ylabel("測試的 Y 軸標籤")
plt.show()

#-------------------------------------------------------------

# 從 CSV 格式的檔案載入資料，並繪製長條圖
import csv
print("\n--------(從 CSV 格式載入資料，並繪製長條圖)--------\n")

# 1. 準備 CSV 資料數據
with open("bar-chart-data.csv", mode="w", encoding="utf-8", newline=""
  ) as file:
  w = csv.writer(file)
  w.writerow(["花費項目", "金額"])
  w.writerow(["飲食", 10000])
  w.writerow(["房租", 12000])
  w.writerow(["娛樂", 5000])
  w.writerow(["交通", 3000])
  w.writerow(["衣裝", 1000])


# 2. 讀取資料
with open("bar-chart-data.csv", mode="r", encoding="utf-8", newline=""
  ) as file:
  r = csv.DictReader(file)
  header = r.fieldnames
  print(f"標頭: {header}")

  x = []
  height = []

  for row in r:
    print(f"花費項目: {row['花費項目']}")
    print(f"金額: {row['金額']}")
    x.append(row["花費項目"])
    height.append(int(row["金額"]))

print(f"\n查看整理後的 x: {x}")
print(f"查看整理後的 height: {height}\n")


# 3. 繪製圖表
plt.figure(figsize=(8, 5))  #將畫布設定為 8x5 ，讓圖片更清楚

plt.bar(x, height, width=0.6, color="green")
plt.xlabel(header[0])
plt.ylabel(header[1])
plt.show()








#=====================================
# Malplotlib長條圖: 資料點的呈現
#=====================================

# X 軸為數字，連續性

# 使用 bar 方法
# 根據資料點畫出長條圖

# 資料點 (x, height) 範例如下:
# (1, 4),(3, 6),(4, 2)
# 程式碼:
# plt.bar([1, 3, 4], [4, 6, 2])

#-------------------------------------------------------------

# X 軸為字串，離散性

# 使用 bar 方法
# 根據資料點畫出長條圖

# 資料點(x, height) 範例如下:
# ("T", 4),("A", 6),("O", 2)
# 程式碼:
# plt.bar(["T", "A", "O"], [4, 6, 2])

#=================================
# 樣式與軸線標題: 圖表的樣式
# 使用 width 參數設定寬度
# 使用 color 參數設定顏色
#=================================

# 資料點 (x, height) 範例如下:
# ("T", 4),("A", 6),("O", 2)
# 程式碼:
"""
plt.bar(
    ["T", "A", "O"],
    [4, 6, 2],
    width=0.5,
    color="red"
)
"""

#====================================
# 軸線的標題
# 使用 xlabel 與 ylabel 方法
# 設定軸線標題
#====================================

# 程式碼:
plt.bar(["T", "A", "O"], [4, 6, 2])
plt.xlabel("X 軸的標題")
plt.ylabel("Y 軸的標題")


# 演練目標
# 熟悉長條圖的操作
# 繪製 X 軸為數字、字串型態的長條圖，觀察差異
# 調整長條圖的樣式與軸線標題
# 載入 CSV 格式的檔案資料，繪製長條圖並加上適當的標示
```
