```python
import matplotlib.pyplot as plt
print("--------(Matplotlib: 散佈圖)--------\n")

# 實際操作: 熟悉散佈圖的操作

# 繪製一組資料點的散佈圖，設定樣式
"""
  一組資料點
  (2, 4), (4, 3), (3, 6)
"""
plt.scatter(
    [2, 4, 3], [4, 3, 6],
    c="#888888", s=100
)
plt.show()

#-------------------------------------------------------------

# 繪製兩組資料點的散佈圖，加上標籤與圖例
"""
  兩組資料點
   第一組 (2, 4), (4, 3), (3, 6)
   第二組 (1, 2), (3, 5), (4, 4)
"""
plt.scatter([2, 4, 3], [4, 3, 6], label="第一組")
plt.scatter([1, 3, 4], [2, 5, 4], label="第二組")
plt.legend();
plt.show()

#-------------------------------------------------------------

# 載入 CSV 格式的檔案資料，繪製適當的散佈圖加標示
import csv
print("\n--------(從 CSV 格式載入資料，並繪製散佈圖)--------\n")

# 1. 準備 CSV 資料數據
with open("scatter-chart-data.csv", mode="w", encoding="utf-8", newline=""
  ) as file:
  w = csv.writer(file)
  w.writerow(["性別", "身高", "體重"])
  w.writerow(["男", 175, 70])
  w.writerow(["男", 168, 60])
  w.writerow(["女", 160, 52])
  w.writerow(["男", 172, 75])
  w.writerow(["女", 152, 48])
  w.writerow(["男", 182, 80])
  w.writerow(["女", 158, 54])
  w.writerow(["女", 170, 60])
  w.writerow(["女", 162, 45])
  w.writerow(["男", 174, 75])
  w.writerow(["男", 170, 60])
  w.writerow(["女", 165, 55])
  w.writerow(["男", 165, 58])
  w.writerow(["男", 178, 75])
  w.writerow(["女", 155, 50])
  w.writerow(["女", 163, 52])

# 2. 讀取資料
with open("scatter-chart-data.csv", mode="r", encoding="utf-8", newline=""
  ) as file:
  r = csv.DictReader(file)
  header = r.fieldnames  #讀取標題
  print(f"標頭: {header}")

  data = {
      "男" : {"x":[], "y":[]},
      "女" : {"x":[], "y":[]}
  }

  for row in r:
    gender = row["性別"]
    data[gender]["x"].append(int(row["身高"]))
    data[gender]["y"].append(int(row["體重"]))
  print(f"\n查看整理後的 data 資料\n{data}\n")

# 3. 繪製圖表
plt.figure(figsize=(8, 5))  #將畫布設定為 8x5 ，讓圖片更清楚

plt.scatter(data["男"]["x"], data["男"]["y"], c="blue", s=40, label="男生")
plt.scatter(data["女"]["x"], data["女"]["y"], c="red", s=40, label="女生")

# 4. 完善圖表資訊
plt.title("男 / 女身高體重散佈圖", fontsize=14)
plt.xlabel("身高 (公分)", fontsize=10)
plt.ylabel("體重 (公斤)", fontsize=10)

# 5. 顯示標籤與圖片
plt.legend();
plt.show()








#=======================================
# Matplotlib 散佈圖: 資料點的呈現
# 基礎的使用
#=======================================

# 使用 scatter 方法
# 根據資料點畫出散佈圖

# 資料點 (x, y) 範例如下:
# (2, 4),(4, 3),(3, 6)
# 程式碼:
# plt.scatter([2, 4, 5], [4, 3, 6])

#-------------------------------------------------------------

# 兩組資料點
# 在同一張圖上畫出兩組資料點

# 資料點 (x, y) 範例如下:
# 第一組 (2, 4),(4, 3),(3, 6)
# 第二組 (1, 2),(3, 5),(4, 4)
# 程式碼:
# plt.scatter([2, 4, 3], [4, 3, 6])
# plt.scatter([1, 3, 4], [2, 5, 4])

#=============================
# 樣式與圖例: 散佈點樣式
#=============================

# 使用 c 參數設定顏色
# 使用 s 參數設定大小

# 資料點 (x, y) 範例如下:
# 第一組 (2, 4),(4, 3),(3, 6)
# 第二組 (1, 2),(3, 5),(4, 4)
# 程式碼:
"""
plt.scatter(
    [2, 4, 3], [4, 3, 6],
    c="green", s=50
)
"""

_="""
plt.scatter(
    [1, 3, 4], [2, 5, 4],
    c="#880000", s=100
)
"""

#================
# 標籤與圖例
#================

# 使用 label 參數設定標籤文字
# 使用 lagend 方法，產生圖例

# 資料點 (x, y) 範例如下:
# 第一組 (2, 4),(4, 3),(3, 6)
# 第二組 (1, 2),(3, 5),(4, 4)
# 程式碼:
"""
plt.scatter(
    [2, 4, 3], [4, 3, 6],
    label="第一組"
)
"""
_="""
plt.scatter(
    [1, 3, 4], [2, 5, 4],
    label="第二組"
)
"""
# 設定標籤後，呼叫 legend 產生圖例
# plt.legend();


# 演練目標
# 熟悉散佈圖的操作
# 繪製一組資料點的散佈圖
# 繪製兩組資料點的散佈圖，並且加上標籤與圖例
# 載入 CSV 格式的檔案資料，繪製散佈圖並加上適當的標示
```
