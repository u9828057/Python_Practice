```python
import numpy as np
import matplotlib.pyplot as plt
print("--------(Matplotlib: 散佈圖)--------\n")

# 1. 準備數據 (模擬實驗室量測數據)
np.random.seed(42)  #固定亂數種子，每次跑的結果都一樣

# 產生 50 個 0 到 10 之間的隨機電壓值
voltage = np.random.uniform(0, 10, 50)

# 產生帶有隨機雜訊的電流數據 (模擬真實量測的誤差)
current = 2*voltage + np.random.normal(0, 2, 50)

# 假設同時記錄了元件的「溫度」 (20度到80度之間)
temperature = np.random.uniform(20, 80, 50)


# 2. 繪製散佈圖
plt.figure(figsize=(8, 5))  #將畫布設定為 8x5，讓圖片更清楚

# 使用 scatter 繪圖，並帶入進階參數
# c=temperature: 讓「點的顏色」隨溫度數值改變
# s=temperature*2: 讓「點的大小」隨溫度數值改變 (乘 2 是為了讓泡泡大一點比較好認)
# alpha=0.7: 設定 70% 的不透明度，讓重疊的點也能看清楚
scatter_plot = plt.scatter(voltage, current,
               c=temperature,
               s=temperature*2,
               alpha=0.7,
               cmap="viridis"
              )

# 3. 完善圖表資訊
plt.title("電子元件特性: 電壓/電流 (附帶溫度資訊)", fontsize=14, color="#FF7744")
plt.xlabel("電壓 (V)")
plt.ylabel("電流 (mA)")
plt.grid(True, linestyle="--", alpha=0.5)

# 4. 【進階技巧】:加入顏色條 (Colorbar)
# 表示右邊的漸層顏色代表什麼意義
cbar = plt.colorbar(scatter_plot)
cbar.set_label("元件溫度 (℃)")

# 5. 儲存並顯示圖片
plt.savefig("電子元件特性 - 電壓_電流.png", dpi=300, bbox_inches="tight")
plt.show()










# §補充技巧

#----------------------------------------------------
# 1. np.random.seed(數值): 掌握隨機的「遙控器」
#----------------------------------------------------
# [用途]: 固定亂數種子，確保每次執行的結果「完全一致」
# [參數]: 括號內的數字可以任意設定 (如: 0, 42, 999 等)
# [場景]：想像在玩《Minecraft（我的世界）》，只要輸入一模一樣的地圖種子碼，生成出來的世界地形就完全一樣

# np.random.seed(42)

#--------------------------------------------------------------
# 2. np.random.uniform(下限, 上限, 產生數量): 均勻分布
#--------------------------------------------------------------
# [用途]: 在設定的範圍內，每個數字出現的「機率完全相等」 (視覺上是平坦的長方形)
# [場景]: 適合模擬沒有特定集中趨勢的變數，例如: 環境溫度在 20～25度之間跳動
# [參數]: low (下限)，high (上限)，size (產生數量)

# temp_uniform = np.random.uniform(20, 25, 50)

#-----------------------------------------------------------------
# 3. np.random.normal(平均值, 標準差, 產生數量): 常態分布
#-----------------------------------------------------------------
# [用途]: 產生著名的「鐘型曲線」，中間數值出現機率最高，越往兩側極端值機率越低
# [場景]: 電子實驗中最重要，非常適合用來模擬「儀器測量誤差」或「電子熱雜訊(Thermal Noise)」
# [參數]: loc (平均值/中心點)，scale (標準差/雜訊大小)，size (產生數量)

# 範例: 模擬用示波器量測 5V 直流電壓，帶有 0.05V 的微小儀器雜訊
# voltage_noise = np.random.normal(5, 0.05, 50)



# 快速總結

# 想確保每次模擬結果都一樣？          → 呼叫 seed()
# 變數在一定範圍內隨機亂跳，沒有中心點？    → 使用 uniform()
# 變數有標準值，但帶有儀器測量誤差或物理雜訊？ → 使用 normal()
```
