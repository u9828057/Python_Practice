```python
# 資料: 程式的基本單位

# 數字
3456
3.5

# 字串: 用來表達任何文字，用雙引號""將文字包起來
"測試中文"
"Hello World"

# 布林值
True
False

# List:    串列: 有順序、可動的列表
[3,4,5]
["Hello","World"]

# Tuple:   元組: 有順序、不可變動的列表
(3,4,5)
("Hello","World")

# Set:    集合: 無順序、可變動且不允許重複元素的列表
{3,4,5}
{"Hello","World"}

# Dictionary 字典: 以鍵值對(Key-Value)形式儲存，可變動；但是Key必須是唯一且不可變動的
{"apple":"蘋果","data":"資料"}

# 變數: 用來儲存資料的自訂名稱
## 變數名稱=資料
data=3
x=3

## print(資料)
print(3)
print(True)
print([1,2,3])
print(x)
#取代舊的資料
x=True
print(x)
x="Hello"
print(x)
x={3,4,5}
print(x)
```

```python
# 數字、字串的運算

# 數字運算
x=3+6
print(x)
x=3*6
print(x)

print("--------(小數除法)--------")
x=3/6    #小數除法
print(x)

print("--------(整數除法)--------")
x=3//6   #整數除法
print(x)
x=7//6
print(x)

print("--------(次方、開根號)--------")
x=3**6   #三的六次方
print(x)
x=2**0.5  #開根號
print(x)

print("--------(取餘數)--------")
x=7%3   #取餘數
print(x)
x=7%8
print(x)

print("--------(將變數中的數字+1)--------")
x=2+3
print(x)
x=x+1
print(x)
x+=1   # x=x+1 #將變數中的數字+1
print(x)

# 字串運算
print("--------(字串運算)--------")
s="Hello"
print(s)

s="Hello"+"World"   #字串串接
print(s)
s="Hello" "World"  #隱式字串拼接
print(s)
s="Hello"*3
print(s)


# 跳脫字元
print("--------(跳脫字元)--------")
"""
常用跳脫字元筆記：
\" : 印出純文字的雙引號
\n : 換行 (Newline)
\t : 縮排 (Tab)
\\ : 印出純文字的反斜線
"""
s="Hell\"o"     #跳脫字元"\": 取消緊接在後面那個字元的「語法功能」，把雙引號變成「純文字」印出，避免字串被提早結束
print(s)
s="Hello\nWorld"   #換行\n
print(s)
s="""Hello
World
"""
print(s)
s="Hello\tWorld"   #縮排\t
print(s)
s="Hello\\World"
print(s)

# 字元索引
print("--------(字元索引)--------")
# 字串會對內部的字元編號(索引)，從0開始計算
s="Hello"
print(s[0])
print(s[2])
print(s[1:4])   #取得開頭是:1 結尾是:4 的字串；包含開頭1但不包含結尾4
print(s[1:])    #從索引1開始，一直取到「字串最尾端」
print(s[:4])    #從「字串最前面」開始取，直到索引4 (不包含4)

# 字串切片進階：步數(Step)
# 語法：s[開頭:結尾:步數]
print("--------(切片與步數)--------")
s="Hello"
print(s[::2])    #步數為2，意思是「每隔一個字元跳著取」(會抓到H,l,o)
print(s[1:4:2])   #開頭索引:1，結尾索引:4 (不包含4)；在範圍內步數為2，跳著取 (會抓到e,l)
print(s[::-1])    #步數為-1，代表「倒退走」。會從最後面倒著取到最前面，是Python中「字串反轉」經典的寫法

# 取得字串長度
print("--------(字串長度)--------")
s="Hello World"
print(len(s))    #取得字串長度: "len()"，中間的空白也算為一個字元
```
