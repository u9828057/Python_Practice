```python
# 資料: 程式的基本單位

# 數字
3456
3.5

# 字串: 用來表達任何文字，用雙引號""將文字包起來
"測試中文"
"Hello World"

# 布林值
True
False

# List:    串列: 有順序、可動的列表
[3,4,5]
["Hello","World"]

# Tuple:   元組: 有順序、不可變動的列表
(3,4,5)
("Hello","World")

# Set:    集合: 無順序、可變動且不允許重複元素的列表
{3,4,5}
{"Hello","World"}

# Dictionary 字典: 以鍵值對(Key-Value)形式儲存，可變動；但是Key必須是唯一且不可變動的
{"apple":"蘋果","data":"資料"}

# 變數: 用來儲存資料的自訂名稱
## 變數名稱=資料
data=3
x=3

## print(資料)
print(3)
print(True)
print([1,2,3])
print(x)
#取代舊的資料
x=True
print(x)
x="Hello"
print(x)
x={3,4,5}
print(x)
```

```python
# 數字、字串的運算

# 數字運算
x=3+6
print(x)
x=3*6
print(x)

print("--------(小數除法)--------")
x=3/6    #小數除法
print(x)

print("--------(整數除法)--------")
x=3//6   #整數除法
print(x)
x=7//6
print(x)

print("--------(次方、開根號)--------")
x=3**6   #三的六次方
print(x)
x=2**0.5  #開根號
print(x)

print("--------(取餘數)--------")
x=7%3   #取餘數
print(x)
x=7%8
print(x)

print("--------(將變數中的數字+1)--------")
x=2+3
print(x)
x=x+1
print(x)
x+=1   # x=x+1 #將變數中的數字+1
print(x)

# 字串運算
print("--------(字串運算)--------")
s="Hello"
print(s)

s="Hello"+"World"   #字串串接
print(s)
s="Hello" "World"  #隱式字串拼接
print(s)
s="Hello"*3
print(s)


# 跳脫字元
print("--------(跳脫字元)--------")
"""
常用跳脫字元筆記：
\" : 印出純文字的雙引號
\n : 換行 (Newline)
\t : 縮排 (Tab)
\\ : 印出純文字的反斜線
"""
s="Hell\"o"     #跳脫字元"\": 取消緊接在後面那個字元的「語法功能」，把雙引號變成「純文字」印出，避免字串被提早結束
print(s)
s="Hello\nWorld"   #換行\n
print(s)
s="""Hello
World
"""
print(s)
s="Hello\tWorld"   #縮排\t
print(s)
s="Hello\\World"
print(s)

# 字元索引
print("--------(字元索引)--------")
# 字串會對內部的字元編號(索引)，從0開始計算
s="Hello"
print(s[0])
print(s[2])
print(s[1:4])   #取得開頭是:1 結尾是:4 的字串；包含開頭1但不包含結尾4
print(s[1:])    #從索引1開始，一直取到「字串最尾端」
print(s[:4])    #從「字串最前面」開始取，直到索引4 (不包含4)

# 字串切片進階：步數(Step)
# 語法：s[開頭:結尾:步數]
print("--------(切片與步數)--------")
s="Hello"
print(s[::2])    #步數為2，意思是「每隔一個字元跳著取」(會抓到H,l,o)
print(s[1:4:2])   #開頭索引:1，結尾索引:4 (不包含4)；在範圍內步數為2，跳著取 (會抓到e,l)
print(s[::-1])    #步數為-1，代表「倒退走」。會從最後面倒著取到最前面，是Python中「字串反轉」經典的寫法

# 取得字串長度
print("--------(字串長度)--------")
s="Hello World"
print(len(s))    #取得字串長度: "len()"，中間的空白也算為一個字元



# ======================================
# 格式化字串 F-String (Python 3.6+)
# ======================================
# 優勢:
# 1.整合性: 將「文字、變數、數學運算」融合在一個字串中，免去繁瑣轉換 "+"號 與 str()
# 2.方便性: 大括號{}內建強大的排版功能(控制小數點、補零)，語法也非常直覺
print("\n\t★ 格式化字串 F-String\n")

# 整合字串與變數
print("--------(F-String: 整合字串與變數)--------")
name="電晶體"
vgs=5
print("這個" + name + "的閘源極電壓為:" + str(vgs) + "V")   # 傳統語法
print(f"這個{name}的閘源極電壓為:{vgs}V")               # F-String語法

item="電阻"
value=100
s="這是一顆" + str(value) + "歐姆的" + item    # 傳統語法
print(s)
s=f"這是一顆{value}歐姆的{item}"            # F-String語法
print(s)


# 括號內即時運算
print("--------(f-string: 括號內即時運算)--------")
v=10
i=2
print("當電壓為"+str(v)+"V、電流為"+str(i)+"A"+"時，此電路的功率為"+str(v*i)+"W")    # 傳統語法
print(f"當電壓為{v}V、電流為{i}A時，此電路的功率為{v*i}W")               # F-String語法

# 大括號{}內支援「動態表達式(Expression)」，可以直接放入數學公式，省去先計算、存成新變數、再印出的繁瑣步驟


# 小數點與排版控制
print("--------(f-string: 小數點與排版控制)--------")
pi=3.1415926535
print(f"圓周率保留兩位小數: {pi:.2f}")
print(f"圓周率保留四位小數: {pi:.4f}")    #最後一位數四捨五入進位了

# 方便性展現：直接在變數後方加上冒號":"來呼叫排版功能
# 限制小數點位數(:.nf)
# f代表浮點數(float)
# .n代表要保留幾位小數點（會自動四捨五入）


# 1. 補零 (通常用在編號)
num=7
print(f"實驗編號: {num:03}")     #總共3位數，不夠的補0
print(f"實驗編號: {num:2<3}")    #總共3位數，不夠的補2，靠左邊對齊
print(f"實驗編號: {num:5>3}")    #總共3位數，不夠的補5，靠右邊對齊
# 想用0以外的字元（或符號）來填補，就必須使用「完整版」的排版語法：{變數: 填充字元 + 對齊方向 + 總長度}

# 2. 對齊 (設定總寬度)
v1=3.2
v2=120.5
print(f"測量值1: {v1:>10}")   #總共10位數，不夠的補空格，靠右對齊
print(f"測量值2: {v2:★<10}")  #總共10位數，不夠的補★，靠左對齊
# > 靠右對齊 (預設)
# < 靠左對齊
# ^ 置中對齊


# 數值進階排版 (處理大型數據或實驗數值非常實用)
print("--------(f-string: 進階數值排版)--------")
# 1. 千位數逗號
resistance=4700000
print(f"這顆電阻的阻抗值為: {resistance:,}")   #自動加上千位逗號，不用自己數零

# 2. 科學記號表示法 (e 或 E)
capacitance=0.0000047
print(f"電容值為: {capacitance:.2e} 法拉")   #變數轉成科學記號e，並保留兩位小數

# 3. 百分比轉換 (%)
efficiency=0.8543
print(f"系統轉換效率為: {efficiency:.1%}")    #把原本的數值乘以100後，加上 % 符號，並保留1位小數
print(f"系統轉換效率為: {efficiency:.2%}")    #把原本的數值乘以100後，加上 % 符號，並保留2位小數
print(f"系統轉換效率為: {efficiency:.0%}")    #把原本的數值乘以100後，加上 % 符號，不保留小數
# 語法拆解：{變數 : .保留位數 %}
# 「:」代表開始執行排版指令
# 「.1」代表小數點後只保留1位 (自動四捨五入)
# 「%」代表自動將小數乘以100，並在最後面補上%符號

# 變數除錯 (Debug)
print("--------(f-string: 變數除錯)--------")
frequency=1500
print(f"frequency = {frequency}")    # 傳統語法
print(f"{frequency = }")       # F-String語法
# 方便性展現：在變數後方加上"="，會自動印出「變數名稱=數值」，抓蟲(Debug)時非常好用


# 資料型態轉換: 整合輸入與輸出
print("--------(文字與數字的轉換)--------")
age_str=input("請輸入你的年齡: ")    #取得輸入: 這時"age_str"的型態是「字串」
age_int=int(age_str)           #型態轉換: 將「字串」轉為「整數」，才能進行計算
next_year=age_int + 1
print(f"你明年就{next_year}歲了!")    #輸出：  計算完畢後，交給f-string負責把數字轉回文字顯示

# f-string只解決了「把數字變成文字印出來（輸出）」的麻煩，但無法解決「把文字變成數字拿來算（輸入）」的問題
# 輸出(Output)：從數字轉文字 ➡️ 交給f-string就好
# 輸入(Input)：從文字轉數字 ➡️ 必須靠int()或float()


# 多行格式化字串 (適合用來產生實驗報告或數據總結)
print("--------(f-string: 多行排版應用)--------")

report =f"""
=============================
    【實驗數據總結報告】
=============================
▶ 測試元件：{name}
▶ 閘源極電壓：{vgs} V
▶ 消耗功率：{v * i} W
▶ 系統轉換效率：{efficiency:.1%}
▶ 訊號頻率：{frequency:,} Hz
=============================
"""
print(report)
```
