```python
# 資料: 程式的基本單位

# 數字
3456
3.5

# 字串: 用來表達任何文字，用雙引號""將文字包起來
"測試中文"
"Hello World"

# 布林值
True
False

# List:    串列: 有順序、可動的列表
[3,4,5]
["Hello","World"]

# Tuple:   元組: 有順序、不可變動的列表
(3,4,5)
("Hello","World")

# Set:    集合: 無順序、可變動且不允許重複元素的列表
{3,4,5}
{"Hello","World"}

# Dictionary 字典: 以鍵值對(Key-Value)形式儲存，可變動；但是Key必須是唯一且不可變動的
{"apple":"蘋果","data":"資料"}

# 變數: 用來儲存資料的自訂名稱
## 變數名稱=資料
data=3
x=3

## print(資料)
print(3)
print(True)
print([1,2,3])
print(x)
#取代舊的資料
x=True
print(x)
x="Hello"
print(x)
x={3,4,5}
print(x)
```
