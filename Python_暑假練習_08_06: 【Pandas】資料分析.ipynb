```python
#=========================================
# 建立與觀察 DataFrame (基礎起手式)
#=========================================
import pandas as pd
print("--------(Pandas: 建立與觀察表格)--------")

# 1. 準備原始資料 (通常會使用 Dictionary 字典來創立)
# 字典的 Key 會變成「欄位名稱」，Value 的陣列會變成「該欄的數據」
experiment_data= {
    "Frequency (Hz)": [10, 100, 1000, 5000, 10000],
    "Gain (dB)": [0.0, -0.1, -3.0, -14.0, -20],
    "Phase (deg)": [0, -5, -45, -78, -90]
}

# 2. 將資料轉換為 DataFrame
df= pd.DataFrame(experiment_data)

# 3. 印出整張表格
print(f"\n{' '*13}完整實驗數據表\n{df}\n")


#========================
# 資料的快速健檢工具
#========================

# .shape: 取得表格的形狀 (列數、欄數)
print(f"資料維度: {df.shape}\n")

# .head(): 只看前 3 筆資料 (適合用來預覽)
print(f"{' '*13}只看前 3 筆資料:\n{df.head(3)}\n")

# .tail(): 只看最後 2 筆資料
print(f"{' '*13}只看最後 2 筆資料:\n{df.tail(2)}\n")

# .describe(): ★ 【實用】一秒算出所有欄位的平均值、標準差、最大與最小值
print(f"{' '*13}所有欄位的統計數據\n{df.describe()}")


#======================================
# 挑選與篩選資料 (Pandas 的強項)
#======================================
import pandas as pd
print("\n--------(Pandas: 資料篩選)--------")

# 情境一: 想單獨拿出「Gain (dB)」這一欄的數據來畫波德圖 (Bode Plot)
gain_data= df["Gain (dB)"]
print(f"\n取得特定欄位「Gain (dB)」的資料\n{gain_data}\n")
#單獨抽出來的一欄，它的型態就會是 Series

# 情境二: 想找出「增益小於或等於 -3dB」的截止頻率區段
cutoff_data= df[df["Gain (dB)"] <= -3]
print(f"{' '*13}截止頻率區段\n{cutoff_data}\n")
#寫法非常直覺，直接把條件塞進中括號裡面
```
