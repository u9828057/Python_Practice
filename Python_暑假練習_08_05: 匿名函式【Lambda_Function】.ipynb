```python
print("--------(Lambda Function: 匿名函式)--------")

# Lambda Function 匿名函式基本練習

# §範例一:

# 一般函式寫法:
# def multiply(n1, n2):
#   return n1*n2

# 匿名函式寫法:
lambda n1,n2: n1*n2
multiply= lambda n1,n2: n1*n2

result= multiply(3, 4)
print(f"乘法的結果: {result}")


# §範例二: 取得現在的時間
from datetime import datetime

# 一般寫法
now_time= datetime.now()
print(f"現在時間 (一般寫法): {now_time}")

# 函式寫法
def get_now():
  return datetime.now()
print(f"現在時間 (函式寫法): {get_now()}")

# 匿名函式寫法
lambda_now= lambda : datetime.now()
print(f"現在時間 (匿名函式寫法): {lambda_now()}")


#==============================
# 搭配 filter() 內建函式
#==============================

ages= [5, 20, 18, 30, 80, 10]
adult_ages= filter(lambda a:a>=18, ages)
adult_ages_list= list(adult_ages)
print(f"成人的年齡: {adult_ages_list}")

#===============================
# 搭配 map() 內建映射函式
#===============================

prices= [100, 50, 200, 75]
discounted_prices=map(lambda p:p*0.8, prices)
print(f"折扣後的價格: {list(discounted_prices)}")






















#=================================================
# 匿名函式: 就是函式，但可以沒有名字
#=================================================

# 匿名函式語法:
# lambda 參數列表:回傳值

# 匿名函式的名稱:
# lambda 參數列表:回傳值
# 變數名稱= lambda 參數列表:回傳值


# 範例一:

def test1(n1, n2):
  return n1 + n2

# 以下匿名函式與上述函式等義
test2= lambda n1, n2: n1+n2
# 呼叫函式
result1= test1(3,4)
result2= test2(3,4)

# 範例二:

def test1():
  return 5*2

# 以下匿名函式與上述函式等義
test2= lambda : 5*2
# 呼叫函式
result1= test1()
result2= test2()

# 範例三:

def test1(*ns):
  return max(ns)*2

# 以下匿名函式與上述函式等義
test2= lambda *ns: max(ns)*2
# 呼叫函式
result1= test1(1,5,-2)
result2= test2(-1,-10,0,-2)


# 不適用的情境

def test(n):
  if n == 0:
    result= "A"
  elif n == 1:
    result= "B"
  else:
    result= "C"
  return result


#=====================================================
# 搭配 filter() 內建函式: 篩選出符合條件的元素
# 語法形式: filter(函式，列表)
#=====================================================

data= [1, 5, -2, 10, -5]

# 篩選 n >= 0 的資料
filtered_data= filter(lambda n:n>=0, data)
filtered_list= list(filtered_data)
print(filtered_list)

#================================
# 搭配 map() 內建函式: 把資料中的每個元素透過某一種規則，統一轉換成另一個資料
# 語法形式: map(函式，列表)
#================================

data= [1,5,-2,10,-5]

# 將 data 中的每個元素都乘上 2，統一轉換成 filtered_map_data 的資料
filtered_map_data= map(lambda n:n*2, data)
filtered_map_list= list(filtered_map_data)
print(filtered_map_list)
```
