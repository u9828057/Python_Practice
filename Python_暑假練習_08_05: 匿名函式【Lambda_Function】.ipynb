```python
print("--------(Lambda Function: 匿名函式)--------")

# Lambda Function 匿名函式基本練習

# §範例一:

# 一般函式寫法:
# def multiply(n1, n2):
#   return n1*n2

# 匿名函式寫法:
lambda n1,n2: n1*n2
multiply= lambda n1,n2: n1*n2

result= multiply(3, 4)
print(f"乘法的結果: {result}")


# §範例二: 取得現在的時間
from datetime import datetime

# 一般寫法
now_time= datetime.now()
print(f"現在時間 (一般寫法): {now_time}")

# 函式寫法
def get_now():
  return datetime.now()
print(f"現在時間 (函式寫法): {get_now()}")

# 匿名函式寫法
lambda_now= lambda : datetime.now()
print(f"現在時間 (匿名函式寫法): {lambda_now()}")


#==============================
# 搭配 filter() 內建函式
#==============================

ages= [5, 20, 18, 30, 80, 10]
adult_ages= filter(lambda a:a>=18, ages)
adult_ages_list= list(adult_ages)
print(f"成人的年齡: {adult_ages_list}")

#===============================
# 搭配 map() 內建映射函式
#===============================

prices= [100, 50, 200, 75]
discounted_prices=map(lambda p:p*0.8, prices)
print(f"折扣後的價格: {list(discounted_prices)}")




















#=================================================
# 匿名函式: 就是函式，但可以沒有名字
#=================================================

# 匿名函式語法:
# lambda 參數列表:回傳值

# 匿名函式的名稱:
# lambda 參數列表:回傳值
# 變數名稱= lambda 參數列表:回傳值


# 範例一:

def test1(n1, n2):
  return n1 + n2

# 以下匿名函式與上述函式等義
test2= lambda n1, n2: n1+n2
# 呼叫函式
result1= test1(3,4)
result2= test2(3,4)

# 範例二:

def test1():
  return 5*2

# 以下匿名函式與上述函式等義
test2= lambda : 5*2
# 呼叫函式
result1= test1()
result2= test2()

# 範例三:

def test1(*ns):
  return max(ns)*2

# 以下匿名函式與上述函式等義
test2= lambda *ns: max(ns)*2
# 呼叫函式
result1= test1(1,5,-2)
result2= test2(-1,-10,0,-2)


# 不適用的情境

def test(n):
  if n == 0:
    result= "A"
  elif n == 1:
    result= "B"
  else:
    result= "C"
  return result


#=====================================================
# 搭配 filter() 內建函式: 篩選出符合條件的元素
# 語法形式: filter(函式，列表)
#=====================================================

data= [1, 5, -2, 10, -5]

# 篩選 n >= 0 的資料
filtered_data= filter(lambda n:n>=0, data)
filtered_list= list(filtered_data)
print(filtered_list)

#================================
# 搭配 map() 內建函式: 把資料中的每個元素透過某一種規則，統一轉換成另一個資料
# 語法形式: map(函式，列表)
#================================

data= [1,5,-2,10,-5]

# 將 data 中的每個元素都乘上 2，統一轉換成 filtered_map_data 的資料
filtered_map_data= map(lambda n:n*2, data)
filtered_map_list= list(filtered_map_data)
print(filtered_map_list)
```

```python
print("--------(Lambda: 匿名函式基礎概念)--------")

#================================
# 1. Lambda的基本語法與差異
#================================

# 傳統寫法
def add_traditional(x, y):
  return x + y
print(f"傳統函式計算寫法: {add_traditional(3, 5)}")

# Lambda的寫法 (匿名的單行函式)
add_lambda= lambda x,y: x+y
print(f"Lambda函式計算寫法: {add_lambda(3, 5)}")

#比較: Lambda 省略了 def、括號、return，強制縮減成一行



#====================================================
#  應用情境一: 自訂排序邏輯 (Data Sorting)
#====================================================
print("\n--------(Lambda 的實際應用情境)--------")

# 假設一組電子學實驗的測量數據，格式為 (頻率 Hz，增益 dB)
experiment_data= [(1000,-20), (10,0), (100,-3)]

# 如果想根據「頻率 (第一個元素)」從小到大排序
# 可以使用 lambda 告知 .sort(): 「請用每筆資料的第 [0] 個位置來排序」
experiment_data.sort(key= lambda data: data[0])
print(f"依照頻率排序: {experiment_data}")

# 如果想根據「增益 (第二個元素)」從小到大排序
experiment_data.sort(key= lambda data:data[1])
print(f"依照增益排序: {experiment_data}")


#=====================================================
#  應用情境二: 搭配 Tkinter (GUI 桌面程式)
#=====================================================
# 這是開發視窗程式最常用到 lambda 的地方
# 在 Tkinter 中，Button 的 command 參數通常不能直接帶入有括號的函式，否則程式一打開就會自動觸發
# 這時就可以用 lambda 來當作「中繼站」

import tkinter as tk

def shutdown_computer(time):
  print(f"系統在 {time} 秒後關機...")

shutdown_computer(5)

# 錯誤寫法: 程式一執行就會立刻印出關機訊息，按鈕會失去作用
# btn= tk.Button(text= "關機", command= shutdown_computer(60))

# 正確寫法: 用 lambda 包裝起來，變成「當按鈕被按下去時，才執行這行函式」
# btn= tk.Button(text= "關機", command= lambda : shutdown_computer(60))


#========================================================
#  應用情境三: 搭配 內建函式 map() 與 filter()
#========================================================

numbers= [1, 2, 3, 4, 5]

# map(): 把串列中每個元素都做同一件事 (如: 把所有數字都平方)
squared_numbers= list(map(lambda n:n**2, numbers))
print(f"平方後的陣列: {squared_numbers}")

# filter(): 篩選出符合條件的元素 (如: 只留下大於 2 的數字)
filtered_numbers= list(filter(lambda n:n>2, numbers))
print(f"篩選後的陣列: {filtered_numbers}")



#=========================================
# 2. Lambda 中的 if-else 條件判斷
#=========================================
print("--------(Lambda 中的 if-else 條件判斷)--------")

# 1. 基礎判斷範例 (回傳 True/False)

# 語法解構: 回傳 True (如果 x 除以 2 的餘數為 0)，否則回傳 False
check_even_bool= lambda x: True if x%2 == 0 else False

print(f"數字 10 是: {check_even_bool(10)}")   #輸出: True
print(f"數字 15 是: {check_even_bool(15)}")   #輸出: False


# 2. 回傳自訂字串範例

# 回傳布林值不夠直覺，可以改成回傳字串
check_even_str= lambda x: "偶數" if x%2 == 0 else "奇數"

print(f"數字 20 是: {check_even_str(20)}")   #輸出: 偶數
print(f"數字 25 是: {check_even_str(25)}")   #輸出: 奇數


# 3. 實際應用: 搭配 map() 批次處理資料

# 假設有一組量測到的電壓數據，只要超過 5V 就是「超載」，5V 以下就是「正常」
voltages= [3.3, 5.2, 4.8, 6.1, 1.2]

# 使用 lambda 搭配 map() 一次檢查所有數據
status= list(map(lambda v: "過載" if v>5.0 else "正常", voltages))

print(f"原始電壓資料: {voltages}")
print(f"狀態檢查結果: {status}")








#========================
# ▲ Lambda 的限制
#========================

# 1. 只能寫「一行」運算式，不能換行
# 2. 不能包含複雜的邏輯 (例如: 不能寫 if...elif...else 區塊，也不能寫迴圈)
# 3. 如果邏輯太複雜，必須使用一般函式寫法 def，為了使用 lambda 而寫出其他人看不懂的程式碼是本末倒置的

# 總結：什麼時候該用 Lambda 呢？
# 1.不想想名字的時候：這個功能太小，取名字很浪費腦力
# 2.需要把函式當作「參數」傳遞時：例如排序的 key、按鈕的 command，或是資料處理的 map/filter
# 3.邏輯只有一行：如果超過一行，需要改用標準的 def
```
