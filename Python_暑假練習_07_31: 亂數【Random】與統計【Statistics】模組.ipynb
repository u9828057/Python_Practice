```python
#=======================
# Random: 亂數模組
#=======================
# 載入模組
import random
print("--------(Random: 亂數模組)--------")

# .choice()、.sample(): 隨機選取
data= random.choice([1,5,6,10,20])     #從列表中隨機選取 1 個資料
print(data)
data= random.sample([1,5,6,10,20], 3)   #從列表中隨機選取 N 個資料 (目前 N = 3)
print(data)


# .shuffle(): 隨機調換順序，類似「洗牌」的概念
data= [1,5,8,20]
random.shuffle(data)
print("隨機調換順序: ",data)


# .random()、.uniform(): 隨機取得亂數
data= random.random()           #取得 0.0 ～　1.0 之間的隨機亂數
print("取得 0～1 間的亂數: ",data)
data= random.uniform(60, 100)       #取得 60 ～ 100 之間的隨機亂數
print("取得 60～100 間的亂數: ",data)


# .randint(a,b): 隨機取得 a 到 b 之間的「整數」 (包含a 與 b)
data= random.randint(1, 10)
print("隨機取得 1 到 10 之間的「整數」: ",data)


# .seed(): 設定亂數種子。只要種子數字相同，每次跑出來的亂數結果就會完全一樣
random.seed(49)
print("亂數種子(49) 的結果: ",random.random())


# .normalvariate(): 取得常態分配亂數

data= random.normalvariate(100, 10)
print("常態分配 【平均數 100、標準差 10】: ",data)
#取得平均數: 100、標準差: 10 的常態分配亂數，得到的資料多數在 90 ～ 110 之間

data= random.normalvariate(150, 20)
print("常態分配 【平均數 150、標準差 20】: ",data)
#取得平均數: 150、標準差: 20 的常態分配亂數，得到的資料多數在 130 ～ 170 之間

data= random.normalvariate(0, 5)
print("常態分配 【平均數 0、標準差 5】: ",data)
#取得平均值: 0、標準差: 5 的常態分配，得到的資料多數在 -5 ～ 5 之間



#===========================
# Statistics: 統計模組
#===========================
# 載入模組
import statistics
print("--------(Statistics: 統計模組)--------")

# 計算平均數、中位數、標準差
import statistics as stat

# .mean(): 計算列表中的平均數
data= stat.mean([1,4,5,8])
print("平均數: ",data)

# .median(): 計算列表中的中位數
data= stat.median([1,2,3,4,5,8,100])
print("中位數: ",data)

# .stdev(): 計算列表中的標準差
data= stat.stdev([1,2,3,4,5,8,10])
print("標準差: ",data)

# .pstdev(): 計算列表中的母體標準差
data= stat.pstdev([1,2,3,4,5,8,10])
print("母體標準差: ",data)


# .stdev(): 【樣本標準差】：除以 n - 1 ,預設用來處理「抽樣」數據，適用於絕大多數的實驗數據分析
# .pstdev(): 【母體標準差】：除以 n ,當數據涵蓋了「所有可能發生的完整群體」時使用















print("--------(額外整理)--------")
#=======================
# Random: 亂數模組
#=======================
# 隨機選取
# .choice(): 從列表中隨機選取 1 個資料
random.choice([0,1,5,8])

# .sample(): 從列表中隨機選取 N 個資料
random.sample([0,1,5,8], 2)            #隨機選取 2 個資料

# 隨機調換順序
import random
# .shuffle(): 將列表的資料「就地」隨機調換順序
data= [0,1,5,8]
random.shuffle(data)

# 隨機亂數
import random
random.random()
random.uniform(0.0, 1.0)      #取得 0.0 ～ 1.0 之間的隨機亂數

# 常態分配亂數
import random
random.normalvariate(100, 10)    #取得平均數 100、標準差 10 的常態分配亂數
#===========================
# Statistics: 統計模組
#===========================

# 計算平均數、中位數、標準差
import statistics
#計算列表中的平均數
statistics.mean([1,4,6,9])

#計算列表中的中位數
statistics.median([1,4,6,9])

#計算列表中的標準差
statistics.stdev([1,4,6,9])

#計算列表中的母體標準差
statistics.pstdev([1,4,6,9])
```
