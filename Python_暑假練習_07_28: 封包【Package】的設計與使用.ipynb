```python
# 主程式

def main():
# 測試 point 模組
  import geometry.point
  result= geometry.point.distance(3,4)

# 測試 line 模組
  print(f"距離: {result}")
  import geometry.line as line
  result=line.slope(1,1,3,3)
  print(f"斜率: {result}")


# Python標準專案實踐: 執行入口判斷
if __name__ == "__main__":
  main()





#==============================================
# 補充技巧: if __name__ == "__main__":
#==============================================
# 這是標準 Python 主程式的標籤
# 意思是：「只有當這個檔案被當成主程式直接執行時，才執行 main()」
# 如果這個檔案是被別人 import 的，裡面的測試程式碼就不會被亂跑
# 能大幅提升程式碼的安全與重複使用性







# =====================================
#  - 專案資料夾
#  - 主程式.py
#  - 封包資料夾
#    - __init__.py
#    - 模組一.py
#    - 模組二.py
#
# 範例
#
#  - 專案資料夾
#  - main.py        #主程式
#  - geometry       #封包資料夾
#    - __init__.py
#    - point.py     #模組一: 針對點的運算的
#    - line.py     #模組二: 針對線的運算的
#
# =====================================
```

```python
# 建立 geometry 資料夾
!mkdir -p geometry
```

```python
# 建立 __init__.py 檔案
%%writefile geometry/__init__.py
```

```python
# 在封包資料夾中建立 point.py 模組
%%writefile geometry/point.py

def distance(x,y):
  return ((x**2) + (y**2))**0.5
```

```python
# 在封包資料夾中建立 line.py 模組
%%writefile geometry/line.py

def len(x1,y1,x2,y2):
  return ((x2-x1)**2 + (y2-y1)**2)**0.5

def slope(x1,y1,x2,y2):
  return (y2-y1) / (x2-x1)
```
