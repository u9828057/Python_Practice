```python
print("--------(網路連線程式、公開資料串接)--------")

# 網路連線
import urllib.request as request
src= "https://www.ntu.edu.tw/"
with request.urlopen(src) as response:
#  data= response.read()               #取得台灣大學網站首頁的原始碼 (HTML、CSS、JS)
  data= response.read().decode("utf-8")       #進一步將位元組解碼為字串
#print(data)


# 串接、擷取公開資料
import urllib.request as request
import json
src="https://data.taipei/api/v1/dataset/" \
  "296acfa2-5d93-4706-ad58-e83cc951863c?scope=resourceAquire"
with request.urlopen(src) as response:
  data= json.load(response)             #利用 JSON 模組處理 JSON 資料格式
print(data)


# 步驟一: 在終端機(螢幕)上，將 '公司名稱' 的列表測試印出

# 1. 深入 JSON 結構，把包含所有公司的「列表List」抓取出來，存入變數 c_list
c_list= data["result"]["results"]

# 2. 把整個列表印出來，觀察結構形式 (測試用)
print(c_list)

# 3. 使用 for 迴圈，把列表裡面的資料逐個拿出來，並且暫時命名為 company (這時 company 是一個單一字典)
for company in c_list:
  #針對每一個被取出的字典，只提取 Key 為 "公司名稱" 的 Value 值並印出
  print(company["公司名稱"])


# 步驟二: 將提取出來的資料儲存成實體文字檔(.txt)

c_list= data["result"]["results"]
with open("c_list.txt", mode="w", encoding="utf-8") as file:
  for company in c_list:
    file.write(company["公司名稱"]+"\n")



#==========================================================
# 額外技巧1: 利用 .keys() 逐步解析 JSON 的結構
#==========================================================
print("--------(利用 .keys() 逐步解析 JSON 的結構)--------")

# 1. 看最外層 'data' 的結構
print(f"【data】 的結構: {data.keys()}")
#輸出結果: dict_keys(['result'])

# 2. 進入 'result' 看下一層的結構
print(f"【result】 的下一層的結構: {data["result"].keys()}")
#輸出結果: dict_keys(['limit', 'offset', 'count', 'sort', 'results'])

# 2.1 'results'是 List 陣列，不是 Dict 字典，所以不能使用 .keys() 來找關鍵字
#print(data["result"]["results"].keys())
#錯誤訊息: AttributeError: 'list' object has no attribute 'keys'
#解決方法: 改用索引值 [0] 先抓出列表中的第一筆資料，再來觀察裡面的結構

# 3. 抓取 'results' List 的第一筆資料，確認欄位名稱
print(f"【results List】 的第一筆資料: {data['result']['results'][0]}")


#=========================================================================
# 額外技巧2: 當網址太長時，想在程式碼中將它拆成兩行以方便閱讀
#=========================================================================
# ☆方法一：使用括號 () (推薦的寫法)
#
# 把多個字串放在括號 () 裡面，Python 會自動把這些相鄰的字串無縫接合起來，中間不會產生任何換行符號
#
# 範例:
# src = ("https://data.taipei/api/v1/dataset/"
#      "296acfa2-5d93-4706-ad58-e83cc951863c?scope=resourceAquire")
#=========================================================================
#
# ☆方法二：使用反斜線 \
#
# 使用反斜線 \ 告訴 Python 「這一行還沒結束，請接續看下一行」
#
# 範例:
# src = "https://data.taipei/api/v1/dataset/" \
#     "296acfa2-5d93-4706-ad58-e83cc951863c?scope=resourceAquire"
#=========================================================================







#=================================================
# 載入模組
# import urllib.request
#
# 下載特定網址資料
# import urllib.request as request
# with request.urlopen(網址) as response:
#   data= response.read()
# print(data)
#=================================================
```

```python
#================
#  題目練習
#================

import json

# 這是一段模擬從氣象局 API 抓下來的 JSON 字串
json_string = """
{
    "status": "success",
    "station_info": {
        "station_id": "C0A980",
        "location": "信義區",
        "elevation": 25.5
    },
    "weather_data": {
        "date": "2026-08-01",
        "records": [
            {"time": "08:00", "temperature": 27.5, "humidity": 65},
            {"time": "12:00", "temperature": 33.1, "humidity": 50},
            {"time": "18:00", "temperature": 29.3, "humidity": 58}
        ]
    }
}
"""

# 將字串轉換為 Python 的字典 (Dictionary)
my_data = json.loads(json_string)

# ---------------- 你的練習區 ----------------

# 任務 1：請提取並印出這個觀測站的「地點」(location)
# 預期輸出：信義區
print("任務 1 輸出：")
# 你的程式碼寫這裡：

print(f"查詢 【my_data】 底下的 Keys: {my_data.keys()}")
print(f"查詢 【station_info】 底下的 Keys: {my_data['station_info'].keys()}")
print(f"正確的輸出答案: {my_data['station_info']['location']}")

# 任務 2：請提取並印出這份資料的「日期」(date)
# 預期輸出：2026-08-01
print("\n任務 2 輸出：")
# 你的程式碼寫這裡：

print(f"查詢 【weather_data】 底下的 Keys: {my_data['weather_data'].keys()}")
print(f"正確的輸出答案: {my_data['weather_data']['date']}")

# 任務 3：請利用 for 迴圈，找出每一次測量的「時間」與「溫度」
# 預期輸出：
# 08:00 的溫度是 27.5 度
# 12:00 的溫度是 33.1 度
# 18:00 的溫度是 29.3 度
print("\n任務 3 輸出：")
# 你的程式碼寫這裡：

print(f"查詢 【records】 底下的 Keys: {my_data['weather_data']['records']}")
record_list= my_data["weather_data"]["records"]
for n in record_list:
  print(f"正確的輸出答案: {n['time']} 的溫度是 {n['temperature']} 度")
```
