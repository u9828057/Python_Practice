```python
#=====================================================
# 1. 建立 DataFrame (通常使用字典 Dictionary)
#=====================================================
import pandas as pd
print("--------(Pandas: DataFrame 雙維度資料)--------")

# 字典的 Key 會變成 「直向的欄位名稱 (Columns)」
# 字典的 Value (陣列) 會變成該欄位的資料

filter_data= {
   "Frequency_Hz": [10, 100, 1000, 1500, 10000],
   "Gain_dB": [0.0, -0.1, -2.5, -3.1, -20.0],
   "Phase_deg": [0, -5, -40, -45, -90]
}

df= pd.DataFrame(filter_data)

# 像 Series 一樣，可以為 DataFrame 貼上「橫向的專屬索引標籤 (Index)」
df.index= ["Test_1", "Test_2", "Test_3", "Test_4", "Test_5"]

print(f"\n{' '*5}【完整低通濾波器量測資料表】\n{df}\n")


#=====================================
# 2. 觀察 DataFrame 的資料屬性
#=====================================
print("--------(DataFrame: 觀察資料屬性)--------\n")

print(f"資料維度 (.shape): {df.shape}")          #輸出 (5, 3) 代表: 5 列 3 欄
print(f"所有欄位名稱 (.columns): {df.columns}")      #輸出欄位索引標籤
print(f"所有列的索引標籤 (.index): {df.index}")      #輸出列的索引標籤


#=====================================================
# 3. 取得資料 (容易搞混的重點: 直向 vs 橫向)
#=====================================================
print("\n--------(DataFrame: 取得特定資料)--------\n")

# 【直向取欄位】: df["欄位名稱"]

# 單獨抽出的一欄，資料型態就會退化成 Series
gain_series= df["Gain_dB"]
print(f"【單獨取出的 Gain_dB 欄位】\n{gain_series}")


# 【橫向取列】: 分為: df.loc[索引名稱] 與 df.iloc[順序]

# .loc[]: Label-location ，使用自訂的索引標籤名稱取資料
print(f"\n【使用 .loc[] 找索引標籤 Test_3 的資料】\n{df.loc['Test_3']}")

# .iloc[]: Integer-location ，用 0, 1, 2... 的絕對順序取資料
print(f"\n【使用 .iloc[] 找第 3 筆的資料】\n{df.iloc[2]}\n")


#============================
# 4. 新增欄位與廣播運算
#============================
import math
print("--------(DataFrame: 新增欄位與運算)--------\n")

# 假設想在表格最右邊新增一欄「角頻率 rad/s」，公式為: W= 2*π*f
# 寫法非常直覺，直接對整個 Frequency_Hz 欄位做數學運算即可

df["Omega_rad"]= 2 * math.pi * df["Frequency_Hz"]
print(f"【新增角頻率後的表格】\n{df}\n")


#======================================
# 5. 條件篩選 (Pandas 實用技巧)
#======================================
print("--------(DataFrame: 條件篩選)--------")

# 實務上常需要找到例如: 「-3dB 截止頻率」的落點
# 語法: df[ 篩選條件 ]

cutoff_filter= df[ df["Gain_dB"] <= -3.0 ]
print(f"\n{' '*10}【增益小於等於 -3dB 的資料區段】\n{cutoff_filter}")
```
