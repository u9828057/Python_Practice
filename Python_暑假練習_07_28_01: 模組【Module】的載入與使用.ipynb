```python
# 建立 geometry.py 檔案
# 會在這台雲端虛擬主機的裡建立geometry.py檔案
%%writefile Module/geometry.py

#在 geometry 模組中定義幾何運算功能
def distance(x1,y1,x2,y2):
  return ((x2-x1)**2 + (y2-y1)**2)**0.5

#定義計算線段的斜率
def slope(x1,y1,x2,y2):
  return (y2-y1) / (x2-x1)

```
