```python
print("--------(NumPy: 多維陣列 ndarray 切割操作)--------")

# 實際操作: 多維陣列的切割操作

# 載入 NumPy
import numpy as np

#-------------------------------------------------------------

# 準備測試資料
arr = np.array([          #2x6
    [2, 4, 6, 8, 10, 12],
    [1, 3, 5, 7, 9, 11]
])

# 根據第一個維度切割
result1 = np.vsplit(arr, 2)
print(f"\n【arr 陣列】\n{arr}\n")
print(f"【切割 第一個維度 後的陣列】\n{result1}")
"""
  1x6
    [[2, 4, 6, 8 ,10, 12]]
  1x6
    [[1, 3, 5, 7, 9, 11]]
"""

# 根據第二個維度切割
result2 = np.hsplit(arr, 2)
print(f"【切割 第二個維度 後的陣列】\n{result2}")
"""
  2x3
    [[2, 4, 6],[1, 3, 5]]
  2x3
    [[8, 10, 12],[7, 9, 11]]
"""

# 根據第二個維度切割成 3 個陣列
result3 = np.hsplit(arr, 3)
print(f"【切割 第二個維度 (三個陣列) 後的陣列】\n{result3}")
"""
  2x2
    [[2, 4],[1, 3]]
  2x2
    [[6, 8],[5, 7]]
  2x2
    [[10, 12],[9, 11]]
"""



#=======================
# 切割 (Splitting)
#=======================

# 切割操作
# 將一個多維陣列，切割成多個

# 根據第一個維度切割
# np.vsplit(陣列, 切割數量)

# 根據第二個維度切割
# np.hsplit(陣列, 切割數量)

#-------------------------------------------------------------

# 切割範例
arr1 = np.array([    #2x4
    [1, 2, 3, 4],
    [5, 6, 7, 8]
])

result1 = np.vsplit(arr1, 2)
#[[1, 2, 3, 4]]  #1x4
#[[5, 6, 7, 8]]  #1x4

result2 = np.hsplit(arr1, 2)
#[[1, 2],[5, 6]]  #2x2
#[[3, 4],[7, 8]]  #2x2
```
