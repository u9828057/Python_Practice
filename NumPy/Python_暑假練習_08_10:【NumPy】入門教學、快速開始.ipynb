```python
import numpy as np
print("--------(NumPy: 入門教學、快速開始)--------")

# 1. 建立陣列 (從 List 轉換)
list_data= [1, 2, 3, 4]
arr= np.array(list_data)


# 2. 常用的陣列生成工具 (預先配置記憶體，效能很好)
zeros_arr= np.zeros(5)         #建立內容全為 0 的一維陣列: [0, 0, 0, 0, 0]
ones_arr= np.ones((2, 3))        #建立 2x3 內容全為 1 的二維陣列
arange_arr= np.arange(0, 10, 2)    #類似迴圈的 range，可設定間隔: [0, 2, 4, 6, 8]

# 【重要】 np.linspace: 在指定的區間內，均勻產生指定數量的點 (畫平滑曲線必備)
# 例如: 在 0 到 1 之間，均勻產生 5 個點
linspace_arr= np.linspace(0, 1, 5)  #[0.0, 0.25, 0.5, 0.75, 1.0]


# 3. 觀察陣列的 3 個關鍵屬性
print(f"\n【陣列的維度數量 .ndim】: {ones_arr.ndim}")
print(f"【陣列的形狀 .shape】: {ones_arr.shape}")
print(f"【陣列的總元素數量 .size】: {ones_arr.size}\n")



#===============================
# 工程實戰: 交流電訊號模擬
#===============================
import numpy as np
print("--------(實務範例: 交流電訊號模擬)--------\n")

# 假設要模擬一個頻率為 10 Hz 的正弦波訊號
# 公式為: V(t)= 5*sin(2πft)
# 設定頻率為 10 Hz
freq= 10

# 1. 建立時間軸 (t): 從 0 到 0.5 秒，取 1000 個連續的取樣點
t= np.linspace(0, 0.5, 1000)


# 2. 向量化運算: 直接將時間陣列 t 帶入公式
# NumPy 會自動對這 1000 個點同時做 sin 運算與乘法
voltage= 5*np.sin(2*np.pi*freq*t)


# 3. 模擬訊號經過濾波器，發生 -3dB 衰減 (電壓值乘上約 0.707)
# §補充技巧: 實務上避免使用 0.707 概算，改用 (1 / np.sqrt(2)) 來確保計算數值的精準度
v_out= voltage * (1 / np.sqrt(2))


# 4. 分析訊號特徵: 抓取波峰 (最大值) 與發生時間
v_max= np.max(v_out)        #取得 v_out 陣列中的最大數值
max_index= np.argmax(v_out)    #取得最大數值在陣列中的「索引位置 (Index)」
t_max= t[max_index]        #利用該索引位置，去 t 陣列查詢對應的時間點


# 結果觀察
print(f"【輸出電壓的取樣點數量】: {v_out.size}個")
print(f"【前三個時間點的輸出電壓】: {v_out[:3]}")
print(f"【最大輸出電壓 (波峰)】: {v_max:.2f}V")
print(f"【波峰發生時間】: {t_max:.3f}秒")
```
