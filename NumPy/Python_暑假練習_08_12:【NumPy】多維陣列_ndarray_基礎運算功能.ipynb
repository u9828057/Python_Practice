```python
print("--------(NumPy: 多維陣列 ndarray 基礎運算功能)--------")

# 實際操作: 多維陣列的各種運算功能

# 載入 NumPy
import numpy as np

#-------------------------------------------------------------

# 逐元運算 (Elementwise)
print("\n--------(Elementwise: 逐元運算)--------")

# 建立兩個形狀相同 (長度為 3) 的一維陣列
data1= np.array([3, 2, 10])
data2= np.array([1, 3, 6])

# 形狀相同，可以順利逐元相加
result= data1 + data2
print(f"\ndata1: {data1}")
print(f"data2: {data2}\n")
print(f"【相加結果】: {result}")
#[3+1, 2+3, 10+6]

# 逐元相乘
result= data1 * data2
print(f"【乘法結果】: {result}")
#[3*1, 2*3, 10*6]

# 逐元比較: 大於 (>)
result= data1 > data2
print(f"【data1 > data2】: {result}")
#[3>1 (True), 2>3 (False), 10>6 (True)]

# 逐元比較: 等於 (==)
result= data1 == data2
print(f"【是否相等】: {result}\n")
#[3==1 (False), 2==3 (False), 10==6 (False)]

#-------------------------------------------------------------

# 矩陣運算 (Matrix)
print("--------(Matrix: 矩陣運算)--------")

# 建立兩個多維陣列
data1= np.array([    #1x2
    [1, 3]
])

data2= np.array([    #2x3
    [2, -1, 3],
    [-2, 4, 1]
])

# 1. 矩陣內積/矩陣乘法 (Dot Product/Matrix Multiplication)
# §數學條件: 第一個矩陣的行必須等於第二個矩陣的列
# 運算過程: (1x2)·(2x3) → (1x3)
result= data1.dot(data2)
# §補充技巧: Python 3.5 後，更推薦使用 data1@data2 這個矩陣乘法符號
print(f"\ndata1: {data1}")
print(f"data2: \n{data2}")
print(f"\n【內積結果】\n{result}\n")

# 2. 張量外積 (Outer Product)
# §數學定義: 兩陣列元素互相交乘，將長度 2 與長度 6 的陣列，展開成 2x6 的大矩陣
result= np.outer(data1, data2)
print(f"【張量外積結果】\n{result}\n")

# 3. 向量外積/叉積 (Cross Product) - 電磁學與物理工程常用
# §數學定義: 計算兩個 3D 向量的法向量
vec1= np.array([1, 0, 0])   #X軸向量
vec2= np.array([0, 1, 0])   #Y軸向量
result_cross= np.cross(vec1, vec2)
print(f"【向量外積 (叉積) 結果】: {result_cross}\n")  #結果為: [0, 0, 1] (Z軸向量)

#-------------------------------------------------------------

# 統計運算 (Statistics)
print("--------(Statistics: 統計運算)--------")

# 建立一個多維陣列
data= np.array([
    [2, 1, 7],  #2x3
    [-5 ,3, 8]
])

# 1. 全域運算 (對整個陣列的所有元素進行統計)

# .sum(): 取得陣列加總
result= data.sum()
print(f"\ndata:\n{data}")
print(f"\n【總和】: {result}")

# .max(): 取得陣列最大值
result= data.max()
print(f"【最大值】: {result}")

# .mean(): 取得陣列平均值
result= data.mean()
print(f"【平均值】: {result:.2f}")

# .std(): 取得陣列母體標準差
result= data.std()
print(f"【母體標準差】: {result:.2f}")

# §補充資料: NumPy 的 .std() 預設為「母體標準差」
# 如果要計算實驗常用的「樣本標準差 (分母為 N-1)」，需要加上 ddof=1 參數
result= data.std(ddof=1)
print(f"【樣本標準差 (ddof=1)】{result:.2f}")


# 2. 指定軸向運算 (使用 axis 參數)

# 針對第一個維度做總和
# axis=0: 針對第一個維度運算 (往下壓縮，最後得到一排每一欄的結果)
result= data.sum(axis=0)  #針對欄做總和 Column (針對第一個維度做總和)
print(f"\n【針對第一個維度 (axis=0) 做總和】: {result}")

# 針對第二個維度做總和
# axis=1: 針對第二個維度運算 (往右壓縮，最後得到一排每一列的結果)
result= data.sum(axis=1)  #針對列做總和 Row (針對第二個維度做總和)
print(f"【針對第二個維度 (axis=1) 做總和】: {result}")

# 針對第一個維度取得最大值
result= data.max(axis=0)
print(f"【針對第一個維度 (axis=0) 找最大值】: {result}")

# 針對第二個維度取得平均值
result= data.mean(axis=1)
print(f"【針對第二個維度 (axis=1) 找平均值】: {result}")


# 3. 累加運算 (Cumulative Sum)
# 實務意義: 常用於計算連續訊號的數值積分、累積能量或累積機率

# 未指定 axis: 陣列會先被攤平成一維，然後逐一往後累加
result= data.cumsum()
print(f"\n【逐值累加】\n{result}")

# axis=0: 針對第一個維度 (垂直往下) 做逐值累加
result= data.cumsum(axis=0)  #針對欄做逐值累加 Column (針對第一個維度做逐值累加)
print(f"【針對第一個維度做逐值累加】\n{result}\n")












#--------------------------------------------------------
# 多維陣列的運算: 基於 ndarray 多維陣列的運算功能
#--------------------------------------------------------

# 逐元運算 (elementwise)
# 矩陣運算 (matrix)
# 統計運算 (statistics)

#-------------------------------------------------------------

# 逐元運算: 針對陣列中的每個資料，逐一進行基本運算

# 建立兩個多維陣列
data1= np.array([3, 4, 1])
data2= np.array([-2, 5, -8])

# 逐一元素運算
data1 + data2
data1 - data2
data1 * data2
data1 / data2
data1 > data2
data1 == data2
data1 <= data2

#-------------------------------------------------------------

# 矩陣運算: 針對兩個陣列進行矩陣運算

# 建立兩個多維陣列
data1= np.array([    #1x2
    [2,1]
])

data2= np.array([    #2x3
    [3, 2, 0],
    [3, 1, -1]
])

# 矩陣運算
data1.dot(data2)      #內積 1x3
data1@data2        #內積 1x3
np.outer(data1, data2)  #外積 2x6

#-------------------------------------------------------------

# 統計運算: 針對單一陣列進行統計運算

# 建立一個多維陣列
data= np.array([
    [2, 1, 7],
    [3, -5, 8]
])

# 統計運算
data.sum()       #全部加總
data.sum(axis= 0)   #加總欄位 (Column)
data.sum(axis= 1)   #加總列  (Row)
data.min()       #全部最小值
data.cumsum()     #逐值累加
data.mean()      #平均數
data.std()       #母體標準差 (NumPy預設)
data.std(ddof=1)    #樣本標準差 (工程實驗常用)
```

```python
import numpy as np
print("--------(NumPy: 多維陣列 ndarray 基礎運算功能)--------")

#==============================================
# 1. 逐元運算 (Element-wise Operations)
# 這是 NumPy 最直覺的運算方式，當對陣列進行加減乘除時，它會「自動對齊」裡面的每一個元素，逐一進行計算
#==============================================

# A. 陣列與單一數字(純量)的運算
print("\n--------(逐元運算: 陣列與單一數字(純量)的運算)--------")

# §假設有一個感測器蒐集到的電壓陣列，因為訊號太弱，需要將所有的數值放大兩倍
voltage= np.array([1.5, 2.0, 3.1])
# 直接將整個陣列乘上 2 倍
amp_voltage= voltage * 2
print(f"\nvoltage: {voltage}")
print(f"\n【放大兩倍的陣列數值】: {amp_voltage}")


# B. 陣列與陣列的運算
print("\n--------(逐元運算: 陣列與陣列的運算)--------")

# §工程應用情境: 計算電功率
# 在電子學中，功率 P = V*I，如果有一個電壓陣列 V 和一個電流陣列 I，在 NumPy 裡完全不需要寫 for 迴圈，可以直接相乘
V= np.array([5, 10, 15])  #電壓
I= np.array([2, 2, 3])   #電流

# 逐一對應相乘
P= V*I
print(f"\nV: {V}")
print(f"I: {I}\n")
print(f"【計算電功率】: {P}")
#輸出: [10 20 45]   (5*2, 10*2, 15*3)



#================================================
# 2. 統計與聚合運算 (Aggregate Functions)
# 在工程數據分析中，常需要知道整批資料的平均值、最大值或總和，在交流電模擬中的 np.max() 就屬於此類
#================================================
print("\n--------(統計與聚合運算: 實驗模擬情境)--------")

# 實驗模擬情境: 假設用示波器擷取了一個 RC 充電電路的 5 個電壓採樣點 (單位: mV)
voltage_data= np.array([10, 20, 30, 40, 50])

# .sum(): 取得陣列加總
print(f"\n【電壓總和】: {voltage_data.sum()}")
#實務意義: 可用於估計累計能量

# .mean(): 取得陣列平均值
print(f"【電壓平均值】: {voltage_data.mean()}")
#實務意義: 可用於求出該段時間內的 DC Offset 或平均功耗

# .max(): 取得陣列最大值
print(f"【電壓最大值】: {voltage_data.max()}")
#實務意義: 可用於抓取峰值電壓 (Vpeak)，確認電路是否會因為瞬間突波而燒毀

# .min(): 取得陣列最小值
print(f"【電壓最小值】: {voltage_data.min()}")


# §工程計算提醒: NumPy 的 .std() 預設為「母體標準差」
print(f"【電壓標準差 (母體)】: {voltage_data.std():.3f}")
#實務意義: 可用於評估訊號的「雜訊波動程度 (Noise Ripple)」

# 標準差越大，代表電壓越不穩定，如果要計算量測儀器抽樣的「樣本標準差」，需要加入參數 ddof=1
print(f"【電壓標準差 (樣本)】: {voltage_data.std(ddof=1):.3f}")



#=======================================================
# 3. 數學通用函式 (Universal Functions, ufuncs)
# NumPy 內建了大量的數學函式，可以直接套用在整個陣列上，不需寫 for 迴圈
#=======================================================
print("\n--------(數學通用函式: 電子實驗情境)--------")

# 實驗情境 1: 交流電壓訊號的相位取樣 (np.sin())

# 假設擷取了一個正弦波訊號，在 0度、90度、180度 的瞬間狀態
phases= np.array([0, np.pi/2, np.pi])

# 算出對應的電壓振幅比例
ac_voltage= np.sin(phases)
print(f"\n【交流電相位振幅 (0度, 90度, 180度)】: {ac_voltage}")
#實務意義: 可用於驗證正弦波在 90 度時達到峰值 (1.0)，在 180 度交越零點
# §補充資訊：180 度的 sin 值印出來是 1.22e-16，是因為電腦浮點數計算的極限，實務上會將它視為 0

#-------------------------------------------------------------

# 實驗情境 2: RC 電容充放電暫態響應 (np.exp())

# 假設時間常數 τ = RC = 1，觀察時間 t=0, 1, 5秒時，電壓衰減的比例 (e**(-t))
t= np.array([0, 1, 5])

# 一次算出三個時間點的自然指數衰減值
decay_ratio= np.exp(-t)
print(f"【RC 暫態電壓衰減比例 ( t= 0, 1, 5τ)】: {decay_ratio}")
#實務意義: 可用於驗證經過 1 個 τ時，電壓剩下約 36.8%；經過 5個 τ時，電壓趨近於 0 (視為完全放電)

#-------------------------------------------------------------

# 實驗情境 3: 主動式低通濾波器的截止頻率 (np.sqrt())

# 在頻率響應圖 (Bode plot) 中，當輸出功率剩下最大值的一半時，電壓衰減為 1/√2
v_ratio= 1 / np.sqrt(2)
print(f"【-3dB 頻率點的電壓衰減值】: {v_ratio:.3f}\n")
#實務意義: 可用於搜尋半功率點 (-3dB Cutoff Frequency)，也就是訊號衰減至約 0.707 倍的關鍵頻率
```
