```python
#=======================================
# List: 有順序、可變動的列表
#=======================================
print("--------(List: 有順序、可變動的列表)--------")
grades=[12,60,15,70,90]
print(grades)
print(grades[0])
grades[0]=55         #把55放到列表"grades"中的第一格位置
print(grades)
print(grades[0])

print(grades[1:4])      #取得列表中的特定數值，取開頭:1 結尾:4 (不包含4)
grades[1:4]=[]        #連續刪除列表中的特定數值，取開頭:1 結尾:4 (不包含4)
print(grades)

grades=grades+[12,33]    #列表串接
print(grades)

print(len(grades))     #取得列表長度 (方法一)
length=len(grades)     #取得列表長度 (方法二)
print(length)

# 巢狀串列 (Nested List)
print("--------(Nested List: 巢狀串列)--------")
data=[[3,4,5],[6,7,8]]
print(data[0][2])                 #先抓取第0個大群組[3,4,5]，再從裡面抓取第2個位置的數字

#利用切片替換局部內容 (替換的數量可以不一樣)
data[0][0:2]=[5,5,5]               #將第0個群組中，開頭0到結尾2(不含2)的元素，轉換為[5,5,5]
print(data)


#================================
# List的專屬工具 (內建方法)
#================================
print("--------(List: 常用內建方法)--------")
my_list=[10,20,30]
print(f"原始列表: {my_list}")       #輸出: [10,20,30]

# 1.append(): 在列表的「最尾端」新增一個元素 (非常常用)
my_list.append(40)
print(f"append之後: {my_list}")      #輸出: [10,20,30,40]

# 2.insert(): 在指定的「索引位置」插入新增元素
my_list.insert(1,15)            #在索引1 的位置，安插數字 15
print(f"insert之後: {my_list}")      #輸出: [10,15,20,30,40]

# 3.remove(): 刪除「指定數值」 (只會刪除找到的第一個數值)
my_list.remove(20)             #把數值 20 刪除
print(f"remove之後: {my_list}")      #輸出: [10,15,30,40]

# 4.pop(): 拔除並取出「指定位置」的元素 (預設是最後一個元素)
last_item=my_list.pop()          #把最後一個數值 40 取出
print(f"pop之後: {my_list}")       #輸出: [10,15,30]
print(f"被取出的元素是: {last_item}")   #輸出: 40

# 5.sort(): 將列表內的數值進行排序
scores=[85,42,99,60]
scores.sort()               #預設為「由小到大」排序 (遞增)
print(f"sort之後: {scores}")       #輸出: [42,60,85,99]
scores.sort(reverse=True)         #設定為「由大到小」排序 (遞減)
print(scores)               #輸出: [99,85,60,42]

# 6.in: 檢查某個數值是否存在於列表中
print("--------(List: 檢查元素是否存在)--------")
my_list=[10,20,30]
print(20 in my_list)          #輸出: True ,元素20存在於列表中
print(40 in my_list)          #輸出: False ,元素40不存在於列表中

# 7.count() 與 index(): 尋找與統計工具(List、Tuple皆可使用)
print("--------(尋找與統計工具)--------")
data_list=[10,20,30,20,20,50]

#count(): 計算某個數值在列表中出現的「次數」
print(f"數字 20 出現了 {data_list.count(20)} 次")

#index(): 找出某個數值「第一次出現的索引位置」
print(f"數字 30 的位置在索引: {data_list.index(30)}")


#=======================================
# Tuple: 有順序、不可變動的列表
#=======================================
print("--------(Tuple: 有順序、不可變動的列表)--------")
data=(3,4,5)
#data[0]=5    #錯誤:Tuple的資料不可以變動
print(data)

# Tuple技巧: 多變數拆解(Unpacking)
print("--------(Tuple: 拆解賦值(Unpacking))--------")
coordinates=(120.5,23.5)
x, y=coordinates             #把Tuple裡面的數值，依序分配給 x 與 y
print(f"經度x= {x}, 緯度y= {y}")

# Tuple陷阱: 單一元素的寫法
print("--------(Tuple: 單一元素陷阱)--------")
#如果 Tuple 只有一個元素，"必須"在後面加上逗號，否則會被 Python 當成普通的數學括號
wrong_tuple=(5)
print(wrong_tuple)        #被當成整數5
print(type(wrong_tuple))
correct_tuple=(5,)
print(correct_tuple)       #真正的 Tuple
print(type(correct_tuple))

```

```python
# 集合的運算
print("--------(set: 集合的運算)--------")
s1={3,4,5}
print(3 in s1)
print(10 in s1)
print(10 not in s1)
s1={3,4,5}
s2={4,5,6,7}

s3=s1&s2     #交集 & : 抓取兩個集合中，相同的資料
print(s3)

s3=s1|s2     #聯集 | : 抓取兩個集合中的所有資料，但不重複抓取元素
print(s3)

s3=s1-s2     #差集 - : 從 s1 中，減去和 s2 重疊的部分
print(s3)

s3=s1^s2     #反交集 ^ : 抓取兩個集合中，不重疊的部分
print(s3)

s=set("Hello")  #set(字串): 把字串中的字母拆解成集合
print(s)
print("H" in s)
print("A" in s)

# 集合的單一資料變動
print("--------(set: 新增與移除)--------")
my_set={3,4,5}

# 1.add(): 新增元素(如果元素已經存在，集合不會有變化)
my_set.add(6)
print(f"新增後: {my_set}")    #輸出: {3,4,5,6}

# 2.discard(): 安全移除元素
my_set.discard(3)
print(f"移除後: {my_set}")    #輸出: {4,5,6}

##💡:Python也有 remove() 可以用，但如果用 discard() 刪除一個「不存在」的數字，程式不會報錯當機，比較安全


# 字典的運算: Key-Value 配對
print("--------(dict: 字典的運算)--------")
dic={"apple":"蘋果","bug":"蟲蟲"}
print(dic["apple"])
dic["apple"]="小蘋果"
print(dic["apple"])
print("apple" in dic)         #判斷 key 是否存在
print("test" not in dic)

del dic["apple"]            #刪除字典中的鍵值對(Key-Value pair)
print(dic)

dic={x:x*2 for x in [3,4,5]}    #從列表的資料產生字典
print(dic)

# get(): 字典的安全取值法
print("--------(dict: 安全取值 get)--------")
dic={"apple":"蘋果","bug":"蟲蟲"}

# 1.找不到 key 時，預設會回傳 None (空值)，而不會讓程式當機
#print(dic["banana"])                #錯誤
print(dic.get("banana"))              #輸出: None

# 2.也可以自己設定「找不到時的備用文字」
print(dic.get("banana", "★字典沒有這個字"))   #輸出: ★字典沒有這個字

#================================================
# 字典拆解工具: keys(), values(), items()
#================================================
print("--------(dict: 拆解字典)--------")
dic={"apple":"蘋果","bug":"蟲蟲"}

# 1.keys(): 只把所有的 「標籤(key)」 抓取出來
print(list(dic.keys()))              #輸出: ['apple', 'bug']

# 2.values(): 只把所有的 「內容(Value)」 抓取出來
print(list(dic.values()))             #輸出: ['蘋果', '蟲蟲']

# 3.items(): 把 Key 與 Value 包裝成 Tuple ，兩兩一組抓取出來
print(list(dic.items()))             #輸出: [('apple', '蘋果'), ('bug', '蟲蟲')]



#========================================================================
#List(串列) []：有順序、可變動   ➡️ 適合用來裝連續的測量數值或波形數據。

#Tuple(元組) ()：有順序、不可變  ➡️ 適合用來裝絕對不能被改動的設定值或座標。

#Set(集合) {}：無順序、去重複   ➡️ 適合用來過濾重複的雜訊、比對兩組數據的差異。

#Dict(字典) {K:V}：查標籤、配對值 ➡️ 適合用來建立電路元件參數表、實驗設定檔。
#========================================================================
```
