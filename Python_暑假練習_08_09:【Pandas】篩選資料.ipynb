```python
print("--------(Pandas: 篩選資料)--------")

# 載入 Pandas 模組
import pandas as pd

# 篩選練習: Series

s_data= pd.Series([30, 15, 20])
condition= [True, False, True]
filtered_data= s_data[condition]
print(f"\n依照布林值篩選後的資料\n{filtered_data}\n")

s_data= pd.Series([30, 15, 20])
condition= s_data > 18
filtered_data= s_data[condition]
print(f"篩選條件為:\n{condition}")
print(f"依照是否大於 18 篩選後的資料\n{filtered_data}\n")

s_data= pd.Series(["您好", "Python", "Pandas"])
condition= [False, True, False]
filtered_data= s_data[condition]
print(f"依照布林值篩選後的資料\n{filtered_data}\n")

s_data= pd.Series(["您好", "Python", "Pandas"])
condition= s_data.str.contains("P")
filtered_data= s_data[condition]
print(f"依照是否包含字串 'P' 篩選後的資料\n{filtered_data}\n")


# 篩選練習: DataFrame

df= pd.DataFrame({
    "name": ["Amy", "Bob", "Charles"],
    "salary": [30000, 50000, 40000]
})

print(f"先將 DataFrame 的資料印出來預覽\n{df}\n")

condition= [False, True, False]
filtered_data= df[condition]
print(f"依照布林值篩選後的資料\n{filtered_data}\n")

condition= df["salary"] >= 40000
filtered_data= df[condition]
print(f"依照薪水是否大於等於 40000 篩選後的資料\n{filtered_data}\n")

condition= df["name"] == "Amy"
filtered_data= df[condition]
print(f"篩選條件為:\n{condition}\n")
print(f"依照名稱是否為 'Amy' 篩選後的資料\n{filtered_data}\n")



#=================================
# Pandas 資料分析: 篩選資料
#=================================

# Series 篩選資料

# 基本邏輯

# 建立 Series 資料
# import pandas as pd
# data= pd.Series(列表)

# 建立篩選條件 (與資料數量對應的布林值)
# condition= [True, False, True]     #True 代表想要的資料； False 代表不要的資料
# filtered_data= data[condition]


# 常見操作

# 建立 Series 資料
# import pandas as pd
# data= pd.Series(列表)

# 建立篩選條件 (直接透過比較運算產生)
# condition= data > 5
# 根據條件完成篩選
# filtered_data= data[condition]

#-------------------------------------------------------------

# DataFrame 篩選資料

# 基本邏輯

# 建立 DataFrame 資料
# import pandas as pd
# data= pd.DataFrame(字典)

# 建立篩選條件 (與資料列數量對應的布林值)
# condition= [True, False, True]
# 根據條件完成篩選
# filtered_data= data[condition]


# 常見操作

# 建立 DataFrame 資料
# import pandas as pd
# data= pd.DataFrame(字典)

# 建立篩選條件 (透過特定欄位的比較運算產生)
# condition= data[欄位名稱] > 5
# 根據條件完成篩選
# filtered_data= data[condition]
```
