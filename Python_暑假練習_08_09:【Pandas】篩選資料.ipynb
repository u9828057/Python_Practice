```python
#========================================
# 1. 準備實驗數據 (電子零件品管表)
#========================================
import pandas as pd
print("--------(Pandas: DataFrame 資料篩選)--------")

df= pd.DataFrame({
    "Component": ["R1_Resistor", "C1_Capacitor", "R2_Resistor", "L1_Inductor", "R3_Resistor"],
    "Value": [100, 0.1, 220, 1.5, 330],
    "Unit": ["Ohm", "uF", "Ohm", "mH", "Ohm"],
    "Error_pct": [1.2, 5.5, 0.8, 2.1, 6.0],
    "QC_Status": ["Pass", "Fail", "Pass", "Pass", "Warning"]
})

print(f"\n{' '*14}原始電子零件品管表\n{df}\n")


#===========================================================
# 2. 單一條件篩選 (比較運算符號: >， <， ==， !=)
#===========================================================
print("--------(情境 1: 單一數值條件)--------")

# 目標: 找出誤差值 (Error_pct) 大於 2% 的零件
condition_1= df["Error_pct"] > 2.0
# 把條件塞回 df[]裡面，就會濾出結果
filtered_data_1= df[condition_1]

print(f"\n{' '*13}誤差值大於 2% 的零件\n{filtered_data_1}\n")


#=============================================
# 3. 複合條件篩選 (AND 交集、OR 聯集)
#=============================================
print("--------(情境 2: 複合多個條件)--------")

# 目標: 找出「是電阻 (Ohm)」 且 「誤差大於 1%」 的零件
condition_2= (df["Unit"] == "Ohm") & (df["Error_pct"] > 1.0)
filtered_data_2= df[condition_2]

print(f"\n{' '*4}「是電阻(Ohm)」且「誤差大於1%」的零件\n{filtered_data_2}\n")


# §補充資料: Pandas 小陷阱
# 1.不能用 Python 原生的 and 、 or，必須使用 '&' (AND) 、 '|' (OR)
# 2.每一個獨立的條件「必須」用小括號 () 包起來


#========================================
# 4. 字串特徵篩選: .str.contains()
#========================================
print("--------(情境 3: 篩選特定字串)--------")

# 目標: 找出零件名稱裡面包含 "Resistor" 的資料
condition_3= df["Component"].str.contains("Resistor", case=False)
filtered_data_3= df[condition_3]
print(f"\n{' '*6}零件名稱裡面包含 'Resistor' 的資料\n{filtered_data_3}\n")


# §補充資料:
# Pandas 的 .contains() 有一個隱藏參數叫做 case=False（不區分大小寫）
# 只要加上 case=False，不管是大寫、小寫、還是大小寫混雜，全部都能精準抓取


#================================
# 5. 特定名單篩選: .isin()
#================================
print("--------(情境 4: 篩選特定名單)--------")

# 目標: 找出狀態是 Fail 或 Warning 的零件 (假設有很多種情況)
# 當條件有很多個時 (如: == A 或 == B 或 == C)，使用 .isin() 最快速
condition_4= df["QC_Status"].isin(["Fail", "Warning"])
filtered_data_4= df[condition_4]

print(f"\n{' '*10}狀態是 Fail 或 Warning 的零件\n{filtered_data_4}\n")
```
