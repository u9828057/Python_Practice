```python
# 文字檔案的讀取與儲存

#==============================
# 練習一：寫入與讀取純文字
#==============================
print("--------(寫入與讀取純文字)--------")

file= open("data.txt",mode="w")     #開啟檔案
file.write("Hello file")         #操作
file.close()               #關閉檔案

file= open("data.txt",mode="w",encoding="utf-8")     #開啟檔案
file.write("測試中文\n好棒")                #操作
file.close()                       #關閉檔案

# 最佳實務寫法 (寫入)
# 打開名為 data.txt 的檔案，並將產生的「檔案物件」命名為 file
with open("data.txt",mode="w",encoding="utf-8") as file:
  file.write("測試中文\n好棒")

# 讀取檔案
with open("data.txt",mode="r",encoding="utf-8") as file:
  data= file.read()                       #讀取檔案的資料，放入 data 中
print(data)                            #印出 data


#================================
# 練習二：讀取數字並計算總和
#================================
print("--------(讀取數字並計算總和)--------")

# 手動用程式產生一個包含數字的檔案
with open("numbers.txt",mode="w",encoding="utf-8") as file:
  file.write("15\n")
  file.write("25\n")
  file.write("40\n")

# 把檔案中的數字資料，一行一行的讀取出來，並計算總合
total=0
with open("numbers.txt",mode="r",encoding="utf-8") as file:
  for line in file:         #一行一行的讀取
    total+=int(line)
print(f"數字的總合為: {total}")    #總合為: 80 (15+25+40)


#================================
# 使用 JSON 格式讀取、複寫檔案
#================================
print("--------(使用 JSON 格式讀取、複寫檔案)--------")

# 1.準備要寫入的資料 (這是一個Python的 Dictionary)
print("\n--------(步驟一：建立 JSON 檔案)--------")
import json
data= {
    "name": "MyAwesomeApp",
    "version": "1.0.0"
}


# 2.將資料寫入成 config.json 檔案
print("\n--------(步驟二：將資料寫入成 JSON 檔案)--------")
with open("config.json",mode="w",encoding="utf-8") as file:
  json.dump(data, file, indent=4)
#使用 json.dump() 把 Python 字典轉換成 JSON 格式並存進檔案
print(f"檔案: config.json 建立成功!\n")


# 3.讀取 JSON 檔案
print("--------(步驟三：執行讀取程式)--------")
with open("config.json",mode="r",encoding="utf-8") as file:
  data= json.load(file)

print(data)
print("名稱: ", data["name"])
print("版本: ", data["version"])



#====================================================
# 補充練習一: 使用 "a" 模式附加資料 (Append)
#====================================================
print("--------(附加模式 Append)--------")

# 假設這是第一次測量資料，使用 "w" 模式寫入 (會覆蓋舊檔或建立新檔)
with open("sensor_log.txt",mode="w",encoding="utf-8") as file:
  file.write("第一次測量:12.5 V\n")

# 假設這是第二次測量資料，使用 "a" 模式寫入 (接續在檔案尾端)
with open("sensor_log.txt",mode="a",encoding="utf-8") as file:
  file.write("第二次測量:12.8 V\n")

# 讀取驗證: 兩次測量結果都完整保存
with open("sensor_log.txt",mode="r",encoding="utf-8") as file:
  data= file.read()
print(data)


#===============================================================
# 補充練習二: 讀取多行文字並存成列表(List) (三種流派比較)
#===============================================================
print("--------(讀取成列表的三種方式)--------")

# 先準備一個包含三行實驗設備名稱的檔案
with open("equipment.txt",mode="w",encoding="utf-8") as file:
  file.write("Active Low-pass Filter\n")
  file.write("Integrator\n")
  file.write("DAC\n")

# 【流派一：傳統基礎寫法】: 使用 .readlines() 一次將所有行讀取出來，並儲存成一個Python列表 (缺點：尾巴會殘留\n)
with open("equipment.txt",mode="r",encoding="utf-8") as file:
  data_readlines= file.readlines()
print(".readlines() 結果: ", data_readlines)
print(".readlines() 印出第二個設備: ", data_readlines[1])


# 【流派二：實務最乾淨寫法】: 使用 .splitlines() 自動切除換行
with open("equipment.txt",mode="r",encoding="utf-8") as file:
  data_split= file.read().splitlines()
print(".splitlines() 的結果: ", data_split)


# 【流派三：實務進階寫法】: 使用列表推導式搭配 .strip()，可擴充性最強
with open("equipment.txt",mode="r",encoding="utf-8") as file:
  data_comprehension= [line.strip() for line in file]
print("列表推導式 結果: ", data_comprehension)
























#=========================================================
# 檔案操作流程: 開啟檔案 > 讀取或寫入 > 關閉檔案
#=========================================================
#
#         基本語法
#
# 檔案物件= open(檔案路徑,mode=開啟模式)
# 開啟模式: r: 讀取模式, w 寫入模式, r+ 讀寫模式, a 附加模式
# 讀取全部文字: 檔案物件.read()

# 一次讀取一行
# for 變數 in 檔案物件:
# 從檔案依序讀取每行文字到變數中

# 讀取 JSON 格式
# import json
# 讀取到的資料= json.load(檔案物件)

# 寫入 JSON 格式
# import json
# json.dump(要寫入的資料,檔案物件)

# 寫入文字
# 檔案物件.write(字串)

# 最佳實務寫法
# with open(檔案路徑,mode=開啟模式) as 檔案物件: 讀取或寫入檔案的程式
# 以上區塊會自動、安全的關閉檔案
#=========================================================




#=========================================================================
# 語法: with open("data.txt", mode="w", encoding="utf-8") as file:
#=========================================================================
# 文字解釋:
# 「使用 with 安全語法，以『寫入模式 (w)』與『UTF-8 編碼』開啟名為 data.txt 的檔案，
# 並將系統產生的『檔案物件』命名為 file。」
#
# with：       建立一個安全操作範圍 （確保程式跑完後，系統會自動安全地關閉檔案，不會佔用記憶體）。
# open("data.txt")： 告訴作業系統，去硬碟裡面找名為 data.txt 的實體檔案。
# mode="w"：     設定開啟目的為「寫入（Write）」。
# encoding="utf-8"： 設定文字編碼方式為 UTF-8，確保中文字不會變成亂碼。
# as file：     把打開檔案後產生的「檔案物件」，在程式碼裡取名為 file。
#=========================================================================
```
